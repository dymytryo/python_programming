# Functions 

Functions are essential building blocks in Python that help you organize code into reusable, modular pieces. 

A function groups a set of statements to perform a specific task. It is defined using the def keyword, followed by the function name, parameters in parentheses, and a colon. The function body is indented.

````python
def greet(name):
    """Return a greeting message."""
    return f"Hello, {name}!"

# Example usage:
print(greet("Dmytro"))
print(greet("Susan"))
print(greet("Jose"))
````

Parameters: 
* `formal` - variables listed in the function definition;
* `actual` (`arguments`) - values passed in the function when it is called;

````python
def add(a, b):
    return a + b

# Here, a and b are formal parameters.
# When calling the function, we provide actual parameters:
result = add(3, 5)
print("Sum is:", result)
````

<div style="background-color:#FAF9F6; color:#283618; padding:10px; border-radius:5px;">
<b>Assignment</b>: Define a function <code>countFirst</code> that counts how many times the first letter of a string appears in the rest of the string. The function must take only one argument and return an integer. Characters are compared case-insensitive.
</div>

````python
def countFirst(string):
    """
    Count how many times the first character of a string `string`
    appears later in the string, case-insensitive.
    Returns 0 if the string is empty.
    """
    if not string:
        return 0

    string = string.lower()
    firstLetter = string[0]
    count = 0

    for char in string[1:]:
        if char == firstLetter:
            count += 1

    return count
````

````python
print(countFirst(""))
print(countFirst("Hello"))  
print(countFirst("Bob"))    
print(countFirst("aAaA"))        
````

<b>Assignment</b>: Define a function called <code>mid</code> that takes three integers as parameters, and returns the one that is greater than one of the numbers and less than the other. Assume three integers are distinct.
</div>

````python
def mid(a, b, c):
    """
    Returns the value that is neither the minimum nor the maximum
    of the three distinct integers a, b, and c.
    """
    # Put the numbers into a list...
    nums = [a, b, c]
    # ...sort them in ascending order...
    nums.sort() # Chapter 10 
    # ...and return the one in the middle.
    return nums[1]
````

````python
print(mid(3, 1, 2))   # 2
print(mid(10, 30, 20))# 20
print(mid(-5, 5, 0))  # 0
````

````python
def mid(a, b, c):
    """
    Returns the middle value of three distinct integers a, b, and c
    (i.e., the one that is greater than one of them and less than the other).
    """
    # Check if a is between b and c
    if (a > b and a < c) or (a > c and a < b):
        return a
    # Check if b is between a and c
    elif (b > a and b < c) or (b > c and b < a):
        return b
    # Otherwise, c must be the middle one
    else:
        return c
````

<div style="background-color:#FAF9F6; color:#283618; padding:10px; border-radius:5px;">
<b>Assignment</b>: Define a function called <code>clampedInput</code> with two parameters, called here <code>lowerBound</code> and <code>upperBound</code>, that takes an integer as input from the user and keeps requesting one until the integer falls within <code>lowerBound</code> and <code>upperBound</code>, inclusively, and finally returns the valid integer.
</div>

````python
def clampedInput(lowerBound, upperBound):
    """
    Prompts the user to enter an integer within the inclusive range
    [lowerBound, upperBound]. Keeps asking until a valid integer is entered,
    then returns that integer.
    """
    while True:
        raw = input(f"Enter an integer between {lowerBound} and {upperBound}: ")
        try:
            value = int(raw)
        except ValueError:
            print("❌ That wasn't a valid integer. Please try again.")
            continue

        if lowerBound <= value <= upperBound:
            return value
        else:
            print(f"❌ {value} is not between {lowerBound} and {upperBound}. Please try again.")
````

````python
# This will keep asking until you type an integer from 10 through 20.
result = clampedInput(10, 20)
print("You entered:", result)
````

# `return`

A function may return a value using the `return` statement. If no return is specified, Python returns `None` by default.  

````python
def max_of_two(num1, num2):
    if num1 > num2:
        return num1
    else:
        return num2

print("Maximum:", max_of_two(10, 20))
````

`void` function is the function that returns no value. 

````python
def no_return():
    pass

print("Result of no_return():", no_return())  # Prints 'None'
````

A function can return multiple values by returning them as a tuple.

````python
def min_and_max(numbers):
    """Return the minimum and maximum of a list of numbers."""
    return min(numbers), max(numbers)

nums = [4, 1, 7, 3, 9]
min_val, max_val = min_and_max(nums)
print("Minimum:", min_val, "Maximum:", max_val)
````

Invoking a function executes its body. You can call functions multiple times with different arguments.

````python
def square(x):
    return x * x

number = 4
print(f"The square of {number} is {square(number)}")
````

<b>Assignment</b>: Define the function printPerceivedTemperature that takes the temperature in degrees Celsius as the only parameter and prints:

- "Cold" if the temperature is below 10;
- "Cool" if the temperature is between 10 and 30;
- "Neutral" if the temperature is between 30 and 32;
- "Warm" if the temperature is between 32 and 43;
- "Hot" if the temperature is above 43.
All lower values are inclusive, while upper values are exclusive.
</div>

````python
def printPerceivedTemperature(temp):
    """
    Prints a description of how the temperature feels.
    
    - "Cold"    for temp < 10
    - "Cool"    for 10 <= temp < 30
    - "Neutral" for 30 <= temp < 32
    - "Warm"    for 32 <= temp < 43
    - "Hot"     for temp >= 43
    """
    if temp < 10:
        print("Cold")
    elif temp < 30:      # temp is at least 10 here
        print("Cool")
    elif temp < 32:      # temp is at least 30 here
        print("Neutral")
    elif temp < 43:      # temp is at least 32 here
        print("Warm")
    else:                # temp is 43 or above
        print("Hot")
````

````python
printPerceivedTemperature(5)    # Cold
printPerceivedTemperature(15)   # Cool
printPerceivedTemperature(30)   # Neutral
printPerceivedTemperature(35)   # Warm
printPerceivedTemperature(50)   # Hot
````

````python
import math

def computeEquation(x):
    """
    Computes: 3 * sqrt(2 + 4*x)
              + 4
              + 4*x
              - 2*x**2
    """
    return 3 * math.sqrt(2 + 4*x) + 4 + 4*x - 2*x**2
````

# `args`

Arguments can be passed by their position or by specifying the parameter names (keyword arguments).

````python
def describe_person(name, age):
    print(f"{name} is {age} years old.")

# Positional arguments:
describe_person("Bob", 30)

# Keyword arguments (order can differ):
describe_person(age=25, name="Alice")
````

Default arguments allow you to specify default values for parameters. If no value is provided during a call, the default is used.

````python
def greet_person(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet_person("Alice"))
print(greet_person("Bob", greeting="Hi"))
````

# Top-Down vs Bottom-Up

Start with the main function, using stubs (placeholders) for lower-level functions. Replace stubs with full implementations gradually.

````python
def main():
    print("Main function started.")
    print_calendar()  # Currently a stub function
    print("Main function ended.")

def print_calendar():
    # Stub: only prints a message
    print("print_calendar stub called.")

main()
````

Develop and test lower-level functions first, then integrate them into the main program.

````python
def add(a, b):
    return a + b

def main():
    result = add(2, 3)
    print("Result of add:", result)

main()
````
