# `for` loop

A `for` loop in Python is used to iterate over a sequence (such as a `list`, `tuple`, `string`) or a range of numbers.   
It is particularly useful when the number of iterations is known ahead of time.

```python
for i in range(4):
    print(i)
```

```python
for i in range(4, 8):
    print(i)
```

<div style="background-color:#FAF9F6; color:#283618; padding:10px; border-radius:5px;">
<b>Assignment</b>: Write a for loop that calculates the average of the numbers below 1000 that are not divisible by 2, 5, and 7. Assign the result to average.
</div>


```python
total = 0
count = 0

# Loop through numbers from 1 up to (but not including) 1000
for i in range(1, 1000):
    # Check if the number is not divisible by 2, 5, or 7
    if i % 2 != 0 and i % 5 != 0 and i % 7 != 0:
        total += i
        count += 1

average = total / count  # Calculate the average

print("The average is:", average)
```

# Nested Loops

Nested loops are loops inside other loops.

```python
# Nested loop example: Printing coordinates for a 1x1 grid
for i in range(1, 2):        # Outer loop for rows (1 to 1)
    for j in range(1, 2):    # Inner loop for columns (1 to 1)
        print(f"({i}, {j})", end=" ")  # Print coordinate in one line
    print()  # Newline after each row
```

```python
# Nested loop example: Printing coordinates for a 2x2 grid
for i in range(1, 3):        # Outer loop for rows (1 to 2)
    for j in range(1, 3):    # Inner loop for columns (1 to 2)
        print(f"({i}, {j})", end=" ")  # Print coordinate in one line
    print()  # Newline after each row
```

```python
# Nested loop example: Printing coordinates for a 3x3 grid
for i in range(1, 4):        # Outer loop for rows (1 to 3)
    for j in range(1, 4):    # Inner loop for columns (1 to 3)
        print(f"({i}, {j})", end=" ")  # Print coordinate in one line
    print()  # Newline after each row
```

# `break` and `continue` 

Use `break` to exit a loop early:

```python
total = 0
for num in range(1, 50):
    total += num
    if total >= 100:
        print(f"Reached {total} with number {num}")
        break
```

Use `continue` to skip certain iterations:

```python
for num in range(1, 10):
    if num in (3, 4):
        continue  # Skip the rest of the loop for these values
    print(num)
```
