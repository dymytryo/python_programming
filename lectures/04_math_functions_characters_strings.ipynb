# 1. Numeric functions

Python's built-in functions are available without an import. Only the names qualified with `math.` require `import math`.

| Name | Where it lives | Import |
|---|---|---|
| `abs`, `max`, `min`, `pow`, `round`, `int` | Built in | None |
| `ord`, `chr`, `len`, `str`, `format`, `print` | Built in | None |
| String methods such as `upper` and `find` | On each `str` object | None |
| `ceil`, `floor`, `sqrt`, `pi` | `math` module | `import math` |

`max(10, 2)` returns the larger argument, `10`.

```python
print(max(10, 2))
```

`min(10, 2)` returns the smaller argument, `2`.

```python
print(min(10, 2))
```

Nested calls are evaluated from the inside out. Here the inner `max` gives `10`, then `min(10, 2)` gives `2`, and the outer `max(2, 8)` gives `8`.

```python
print(max(min(max(10, 2), 2), 8))
```

`pow(base, exponent)` is the built-in function form of `base ** exponent`.

```python
print(pow(min(2, 3), 3))
```

An exact tie at `4.5` goes to the even integer `4`.

```python
print(round(4.5))
```

`5.4` is nearer to `5` than to `6`.

```python
print(round(5.4))
```

An exact tie at `5.5` goes to the even integer `6`.

```python
print(round(5.5))
```

`int(4.5)` discards the fractional part and returns `4`.

```python
print(int(4.5))
```

`int(5.4)` discards the fractional part and returns `5`.

```python
print(int(5.4))
```

`int(5.5)` also discards the fractional part and returns `5`.

```python
print(int(5.5))
```

For a negative float, `int` still truncates toward zero.

```python
print(int(-4.9))
```

With two decimal places, `5.466` rounds up to `5.47`.

```python
print(round(5.466, 2))
```

With two decimal places, `5.463` rounds down to `5.46`.

```python
print(round(5.463, 2))
```

The absolute value of `-3` is its distance from zero, `3`.

```python
print(abs(-3))
```

A positive value is unchanged by `abs`.

```python
print(abs(3))
```

# 5. Comparing strings

Strings compare lexicographically by Unicode code point: compare the first characters; if they tie, continue until a pair differs.

```python
print("Hey there" > "hi")
```

Get the code point for uppercase `H`.

```python
print(ord("H"))
```

Get the code point for lowercase `h`.

```python
print(ord("h"))
```

Compare the two code points to explain the string-order result.

```python
print(ord("H") < ord("h"))
```

Single versus double quotes do not affect comparison. They are delimiters, not part of the value.

```python
print('Python' == "Python")
```

Store the first string; this literal stands in for keyboard input.

```python
s1 = "argentina"
```

Store the second string; uppercase `Z` sorts before lowercase `a`.

```python
s2 = "Zimbabwe"
```

Swap the values when the second string sorts first.

```python
s1, s2 = (s2, s1) if s2 < s1 else (s1, s2)
```

Display the two strings in ascending order.

```python
print("Prints smallest string first:", s1, s2)
```

The shorter prefix `cat` sorts before `catalog`.

```python
print("cat" < "catalog")
```

The reverse comparison is false.

```python
print("catalog" < "cat")
```

## Testing string contents

`isdigit` is true only for a non-empty string made entirely of digit characters.

```python
print("1436".isdigit())
```

A string containing digits is not entirely alphabetic.

```python
print("COSC1436".isalpha())
```

A non-empty string containing only letters is alphabetic.

```python
print("COSCXXXX".isalpha())
```

`isalnum` requires a non-empty string and permits only alphabetic characters and digits.

```python
print("COSC1436".isalnum())
```

Spaces do not prevent the cased letters from being uppercase.

```python
print("COSC XXXX".isupper())
```

The same uppercase letters do not satisfy `islower`.

```python
print("COSC XXXX".islower())
```

Letters make the first string fail the whitespace-only test.

```python
print("COSC XXXX".isspace())
```

A non-empty string containing only spaces passes `isspace`.

```python
print("      ".isspace())
```

# 9. Program: course-section report

**Requirement.** Normalize a course code, validate its letters and digits, calculate the number of class sections needed, and print a compact report.

**Input.** Store a course code with extra surrounding whitespace.

```python
raw_course = "  cosc1436  "
```

**Input.** Store the current enrollment.

```python
enrollment = 27
```

**Input.** Store the capacity of one room.

```python
room_capacity = 12
```

**Normalize.** Remove edge whitespace and convert the course code to uppercase.

```python
course = raw_course.strip().upper()
```

**Split.** Take the first four characters as the subject.

```python
subject = course[:4]
```

**Split.** Take the remaining characters as the course number.

```python
course_number = course[4:]
```

**Validate.** The subject must be four letters and the course number must be four digits.

```python
valid_course = len(subject) == 4 and subject.isalpha() and len(course_number) == 4 and course_number.isdigit()
```

**Calculate.** Round the enrollment-to-capacity ratio upward to get the required sections.

```python
sections_needed = math.ceil(enrollment / room_capacity)
```

**Calculate.** Divide enrollment by the section count to get the average size.

```python
average_section_size = enrollment / sections_needed
```

**Output.** Display the normalized subject and course number.

```python
print(f"{'Course:':<18}{subject} {course_number}")
```

**Output.** Display whether the course code passed validation.

```python
print(f"{'Code valid:':<18}{valid_course}")
```

**Output.** Display the first subject letter's Unicode code point.

```python
print(f"{'First code point:':<18}{ord(subject[0])}")
```

**Output.** Display the enrollment.

```python
print(f"{'Enrollment:':<18}{enrollment}")
```

**Output.** Display the required number of sections.

```python
print(f"{'Sections needed:':<18}{sections_needed}")
```

**Output.** Display the average section size to one decimal place.

```python
print(f"{'Average section:':<18}{average_section_size:.1f}")
```
