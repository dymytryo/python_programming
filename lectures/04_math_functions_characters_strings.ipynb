# Math functions, characters, and strings

# 1. Numeric functions

Python's built-in functions are available without an import. Only names qualified with `math.` require `import math`.

| Name | Where it lives | Import |
|---|---|---|
| `abs`, `max`, `min`, `pow`, `round`, `int` | Built in | None |
| `ord`, `chr`, `len`, `str`, `format`, `print` | Built in | None |
| String methods such as `upper` and `find` | On each `str` object | None |
| `ceil`, `floor`, `sqrt`, `pi` | `math` module | `import math` |

Descriptions state reusable contracts with parameter names. Run the code cells to see results for concrete arguments.

`max(a, b, ...)` returns the largest argument.

```python
print(max(10, 2))
```

`min(a, b, ...)` returns the smallest argument.

```python
print(min(10, 2))
```

Nested calls are evaluated from the innermost call outward.

```python
print(max(min(max(10, 2), 2), 8))
```

`pow(base, exponent)` is the built-in function form of `base ** exponent`.

```python
print(pow(min(2, 3), 3))
```

`round(number)` returns the nearest integer. Exact halfway cases use ties-to-even.

```python
print(round(4.5))
```

```python
print(round(5.4))
```

```python
print(round(5.5))
```

`int(number)` discards the fractional part, truncating toward zero.

```python
print(int(4.5))
```

```python
print(int(5.4))
```

```python
print(int(5.5))
```

```python
print(int(-4.9))
```

`round(number, ndigits)` rounds to the requested number of decimal places.

```python
print(round(5.466, 2))
```

```python
print(round(5.463, 2))
```

`abs(number)` returns a number's magnitude—its distance from zero.

```python
print(abs(-3))
```

```python
print(abs(3))
```

# 5. Comparing strings

Strings compare lexicographically by Unicode code point: compare the first characters; if they tie, continue until a pair differs.

```python
print("Hey there" > "hi")
```

```python
print(ord("H"))
```

```python
print(ord("h"))
```

```python
print(ord("H") < ord("h"))
```

```python
print('Python' == "Python")
```

A conditional expression can order two values: `(b, a) if b < a else (a, b)`.

```python
s1 = "argentina"
```

```python
s2 = "Zimbabwe"
```

```python
s1, s2 = (s2, s1) if s2 < s1 else (s1, s2)
```

```python
print("Prints smallest string first:", s1, s2)
```

When one string is an exact prefix of another, the shorter string sorts first.

```python
print("cat" < "catalog")
```

```python
print("catalog" < "cat")
```

## Testing string contents

Content-testing methods return `False` for an empty string.

| Method | `True` when |
|---|---|
| `text.isdigit()` | Every character is a digit |
| `text.isalpha()` | Every character is alphabetic |
| `text.isalnum()` | Every character is alphabetic or a digit |
| `text.isupper()` | At least one cased character exists and all cased characters are uppercase |
| `text.islower()` | At least one cased character exists and all cased characters are lowercase |
| `text.isspace()` | Every character is whitespace |

```python
print("1436".isdigit())
```

```python
print("COSC1436".isalpha())
```

```python
print("COSCXXXX".isalpha())
```

```python
print("COSC1436".isalnum())
```

```python
print("COSC XXXX".isupper())
```

```python
print("COSC XXXX".islower())
```

```python
print("COSC XXXX".isspace())
```

```python
print("      ".isspace())
```

## Searching strings

Search methods inspect positions, prefixes, suffixes, and counts.

| Method | Returns |
|---|---|
| `text.endswith(suffix)` | Whether `text` ends with `suffix` |
| `text.startswith(prefix)` | Whether `text` starts with `prefix` |
| `text.find(part)` | Lowest matching index, or `-1` |
| `text.rfind(part)` | Highest matching index, or `-1` |
| `text.count(part)` | Number of non-overlapping matches |

```python
print("welcome".endswith("come"))
```

```python
print("welcome".startswith("good"))
```

```python
print("welcome".find("we"))
```

```python
print("welcome".find("me"))
```

```python
print("welcome".find("become"))
```

```python
print("welcome".rfind("e"))
```

```python
print("welcome".count("e"))
```

## Stripping characters

Strip methods remove characters from string edges.

| Method | Removes from |
|---|---|
| `text.strip(chars)` | Both edges |
| `text.rstrip(chars)` | Right edge |
| `text.lstrip(chars)` | Left edge |

Without `chars`, they remove whitespace. With `chars`, they remove any listed character repeatedly—not a fixed prefix or suffix.

```python
print(repr("    welcome    ".strip()))
```

```python
print(repr("    welcome    ".rstrip()))
```

```python
print(repr("    welcome".lstrip()))
```

```python
print("---welcome---".strip("-"))
```

```python
print("www.python.org".lstrip("w."))
```

# 9. Program: course-section report

**Requirement.** Normalize a course code, validate its letters and digits, calculate the number of class sections needed, and print a compact report.

**Algorithm.**

1. Normalize the course code with `strip` and `upper`.
2. Split the subject from the course number with slices.
3. Validate both parts with length and content tests.
4. Calculate sections and average size.
5. Format the report.

```python
raw_course = "  cosc1436  "
```

```python
enrollment = 27
```

```python
room_capacity = 12
```

```python
course = raw_course.strip().upper()
```

```python
subject = course[:4]
```

```python
course_number = course[4:]
```

```python
valid_course = len(subject) == 4 and subject.isalpha() and len(course_number) == 4 and course_number.isdigit()
```

```python
sections_needed = math.ceil(enrollment / room_capacity)
```

```python
average_section_size = enrollment / sections_needed
```

```python
print(f"{'Course:':<18}{subject} {course_number}")
```

```python
print(f"{'Code valid:':<18}{valid_course}")
```

```python
print(f"{'First code point:':<18}{ord(subject[0])}")
```

```python
print(f"{'Enrollment:':<18}{enrollment}")
```

```python
print(f"{'Sections needed:':<18}{sections_needed}")
```

```python
print(f"{'Average section:':<18}{average_section_size:.1f}")
```
