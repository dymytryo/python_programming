# Math functions, characters, and strings

# 1. Numeric functions

Python's built-in functions are available without an import. Only the names qualified with `math.` require `import math`.

| Name | Where it lives | Import |
|---|---|---|
| `abs`, `max`, `min`, `pow`, `round`, `int` | Built in | None |
| `ord`, `chr`, `len`, `str`, `format`, `print` | Built in | None |
| String methods such as `upper` and `find` | On each `str` object | None |
| `ceil`, `floor`, `sqrt`, `pi` | `math` module | `import math` |

`max` returns the largest argument; `min` returns the smallest.

```python
print(max(10, 2))
```

```python
print(min(10, 2))
```

Nested calls are evaluated from the inside out. Here the inner `max` gives `10`, then `min(10, 2)` gives `2`, and the outer `max(2, 8)` gives `8`.

```python
print(max(min(max(10, 2), 2), 8))
```

`pow(base, exponent)` is the built-in function form of `base ** exponent`.

```python
print(pow(min(2, 3), 3))
```

With no second argument, `round` returns the nearest integer. An exact halfway value goes to the even neighbour, so `4.5` becomes `4` and `5.5` becomes `6`.

```python
print(round(4.5))
```

```python
print(round(5.4))
```

```python
print(round(5.5))
```

`int` does not round a float. It truncates toward zero by discarding the fractional part.

```python
print(int(4.5))
```

```python
print(int(5.4))
```

```python
print(int(5.5))
```

```python
print(int(-4.9))
```

A second argument tells `round` how many digits to keep after the decimal point.

```python
print(round(5.466, 2))
```

```python
print(round(5.463, 2))
```

`abs` returns a number's distance from zero, so its result is never negative.

```python
print(abs(-3))
```

```python
print(abs(3))
```

# 2. Characters and Unicode

Python has no separate character type. A character is a `str` whose length is one.

```python
letter = "A"
```

```python
print(type(letter))
```

```python
print(len(letter))
```

Python strings use Unicode. ASCII is the first 128 Unicode code points, from `0` through `127`.

```python
print(ord("A"))
```

```python
print(ord("a"))
```

The built-in `ord` returns the code point of one character. The built-in `chr` performs the reverse conversion. Neither needs an import.

```python
print(ord('a'))
```

```python
print(chr(98))
```

```python
print(ord('A'))
```

A character can make a round trip through its integer code point without changing.

```python
character = "$"
```

```python
code_point = ord(character)
```

```python
print(code_point)
```

```python
print(chr(code_point))
```

`chr` is not limited to ASCII; it can create characters from the wider Unicode set.

```python
print(chr(0x03C0))
```

```python
print(chr(0x1F680))
```

# 3. String literals, escapes, and printing

Single and double quotes create the same `str` value. Choose the outer quote that makes the text inside easiest to read.

```python
single_quoted = 'COSC 1436'
```

```python
double_quoted = "COSC 1436"
```

```python
print(single_quoted == double_quoted)
```

Triple quotes can hold both quote characters without escaping either one.

```python
print('''I made an 'A' in "COSC1436"''')
```

A backslash escapes a quote that would otherwise end the string. `end` replaces the newline that `print` normally writes.

```python
print("\"Knowledge\"", end=" ")
```

```python
print('is the key to \'success\'')
```

Escape sequences represent characters that are awkward to type directly: `\n` is a newline, `\t` is a tab, and `\\` is one literal backslash.

```python
print("first line\nsecond line")
```

```python
print("name\tvalue")
```

```python
print("COSC\\1436")
```

To print a quoted backslash message and then a filename on the next line, escape both the quotation marks and the backslash.

```python
print("\"\\hello there!\"\n04slide_accessible.pptx")
```

Successive calls can stay on one output line by supplying different `end` strings.

```python
print("AAA", end=" ")
```

```python
print("BBB", end="")
```

```python
print("CCC", end="***")
```

```python
print("DDD")
```

A final `print()` with no arguments writes only a newline.

```python
print("left", end=" -> ")
```

```python
print("right", end="")
```

```python
print()
```

# 5. Comparing strings

Strings compare lexicographically by Unicode code point: compare the first characters; if they tie, continue until a pair differs.

```python
print("Hey there" > "hi")
```

Uppercase `H` has a smaller code point than lowercase `h`, which explains that result.

```python
print(ord("H"))
```

```python
print(ord("h"))
```

```python
print(ord("H") < ord("h"))
```

Single versus double quotes do not affect comparison. They are delimiters, not part of the value.

```python
print('Python' == "Python")
```

To print two strings in ascending order, compare them and swap when the second is smaller. The literals stand in for keyboard input; uppercase `Z` sorts before lowercase `a`.

```python
s1 = "argentina"
```

```python
s2 = "Zimbabwe"
```

```python
s1, s2 = (s2, s1) if s2 < s1 else (s1, s2)
```

```python
print("Prints smallest string first:", s1, s2)
```

If one string is an exact prefix of another, the shorter string sorts first.

```python
print("cat" < "catalog")
```

```python
print("catalog" < "cat")
```

## Testing string contents

`isdigit` is true only for a non-empty string made entirely of digit characters.

```python
print("1436".isdigit())
```

`isalpha` requires at least one character and every character must be alphabetic.

```python
print("COSC1436".isalpha())
```

```python
print("COSCXXXX".isalpha())
```

`isalnum` requires a non-empty string and permits only alphabetic characters and digits.

```python
print("COSC1436".isalnum())
```

`isupper` and `islower` inspect only cased letters. Spaces and digits do not make an otherwise uppercase string fail.

```python
print("COSC XXXX".isupper())
```

```python
print("COSC XXXX".islower())
```

`isspace` is true only when the string is non-empty and every character is whitespace.

```python
print("COSC XXXX".isspace())
```

```python
print("      ".isspace())
```

## Searching strings

`startswith` and `endswith` test the beginning or ending without extracting a slice.

```python
print("welcome".endswith("come"))
```

```python
print("welcome".startswith("good"))
```

`find` returns the lowest starting index of a substring. It returns `-1` when no match exists.

```python
print("welcome".find("we"))
```

```python
print("welcome".find("me"))
```

```python
print("welcome".find("become"))
```

`rfind` searches from the right; `count` returns the number of non-overlapping matches.

```python
print("welcome".rfind("e"))
```

```python
print("welcome".count("e"))
```

## Stripping characters

`strip` removes whitespace from both ends, `lstrip` from the left, and `rstrip` from the right. `repr` makes the remaining spaces visible.

```python
print(repr("    welcome    ".strip()))
```

```python
print(repr("    welcome    ".rstrip()))
```

```python
print(repr("    welcome".lstrip()))
```

An argument changes stripping from whitespace to a set of edge characters. It removes characters one by one, not one exact prefix or suffix.

```python
print("---welcome---".strip("-"))
```

```python
print("www.python.org".lstrip("w."))
```

# 8. Formatting output

The built-in `format(item, specifier)` returns a string and needs no import. For a float, `.2f` means fixed-point notation with two digits after the decimal point.

```python
print(format(12.345, ".2f"))
```

A leading width reserves at least that many characters. Numbers are right-aligned by default.

```python
print(format(12345.678910, "10.2f"))
```

```python
print(format(12.345, "10.2f"))
```

`<` left-aligns the result inside the requested width.

```python
print(format(12.3, "<10.3f"))
```

The width is optional when only the numeric precision matters.

```python
print(format(12.345, ".2f"))
```

For strings, `s` selects string formatting and `>` right-aligns within the width.

```python
print(format("python", ">10s"))
```

The same mini-language supports scientific notation with `e` and percentages with `%`.

```python
print(format(57.467657, "10.2e"))
```

```python
print(format(0.53457, "10.2%"))
```

Integer type codes include decimal `d`, hexadecimal `x`, octal `o`, and binary `b`.

```python
print(format(59832, "d"))
```

```python
print(format(59832, "x"))
```

```python
print(format(59832, "o"))
```

```python
print(format(59832, "b"))
```

## F-strings

An f-string begins with `f` and evaluates each expression inside braces at runtime.

```python
gpa = 3.5
```

```python
grade = "B"
```

```python
print(f"GPA {gpa} is {grade} grade")
```

After a colon, an f-string uses the same format specifier as `format`.

```python
price = 9.99
```

```python
quantity = 3
```

```python
print(f"Total: ${price * quantity:>8.2f}")
```

`end` works with f-strings exactly as it does with every other `print` call.

```python
roman = 'X'
```

```python
print(f"Roman numeral '{roman}' is", end=' ')
```

```python
print(10)
```

# 9. Program: course-section report

**Requirement.** Normalize a course code, validate its letters and digits, calculate the number of class sections needed, and print a compact report.

**Input.** The values stand in for keyboard input so the complete notebook runs unattended.

```python
raw_course = "  cosc1436  "
```

```python
enrollment = 27
```

```python
room_capacity = 12
```

**Normalize and split.** Remove edge spaces, make the code uppercase, and slice its parts.

```python
course = raw_course.strip().upper()
```

```python
subject = course[:4]
```

```python
course_number = course[4:]
```

**Validate.** The subject must be four letters and the course number must be four digits.

```python
valid_course = len(subject) == 4 and subject.isalpha() and len(course_number) == 4 and course_number.isdigit()
```

**Calculate.** Ceiling division in the problem domain means one partly filled section still counts as a section.

```python
sections_needed = math.ceil(enrollment / room_capacity)
```

```python
average_section_size = enrollment / sections_needed
```

**Output.** F-strings align the labels and format the average to one decimal place.

```python
print(f"{'Course:':<18}{subject} {course_number}")
```

```python
print(f"{'Code valid:':<18}{valid_course}")
```

```python
print(f"{'First code point:':<18}{ord(subject[0])}")
```

```python
print(f"{'Enrollment:':<18}{enrollment}")
```

```python
print(f"{'Sections needed:':<18}{sections_needed}")
```

```python
print(f"{'Average section:':<18}{average_section_size:.1f}")
```
