# Selection statements

# 1. Selection and Boolean values

- A selection statement chooses which statements run next.
- Its condition must be usable as either `True` or `False`.

```python
temperature = 33
is_freezing = (temperature <= 32)
print(is_freezing)
```

- The `bool` type has exactly two values: `True` and `False`.
- Their capital letters matter.

```python
print(True, type(True))
print(False, type(False))
```

- Lowercase `true` and `false` are ordinary names.
- Reading either before assignment raises `NameError`.

```python
print(true)
```

- Converting a Boolean to `int` gives `1` for `True` and `0` for `False`.

```python
print(int(True))
print(int(False))
```

- `bool(0)` and `bool(0.0)` are `False`.
- Every other numeric value is `True`, including negative values.

```python
print(bool(0), bool(0.0))
print(bool(1), bool(102), bool(-10))
```

- A one-way selection has one optional path.
- Its indented block runs only when the condition is `True`.

```python
is_raining = True
if is_raining:
    print("Take an umbrella")
```

- A two-way selection chooses exactly one of two paths.
- `else` runs when the `if` condition is `False`.

```python
is_raining = False
if is_raining:
    print("Take an umbrella")
else:
    print("No umbrella needed")
```

- A multi-way selection tests conditions from top to bottom.
- Only the first true branch runs.

```python
is_snowing, is_raining = False, True
if is_snowing:
    print("Wear boots")
elif is_raining:
    print("Take an umbrella")
else:
    print("Clear weather")
```

# 2. Relational operators

A relational operator compares two values and produces a Boolean.

| Operator | Meaning |
|---|---|
| `<` | less than |
| `<=` | less than or equal to |
| `>` | greater than |
| `>=` | greater than or equal to |
| `==` | equal to |
| `!=` | not equal to |

```python
left_score, right_score = 82, 90
print(left_score < right_score)
print(left_score <= right_score)
print(left_score > right_score)
print(left_score >= right_score)
print(left_score == right_score)
print(left_score != right_score)
```

- `=` assigns the value on its right to the name on its left.
- `==` compares two values and leaves both values unchanged.

```python
score = 88
print(score)
print(score == 88)
print(score)
```

- Arithmetic expressions produce numbers.
- Relational expressions produce Boolean values.

```python
print(6 + 4, type(6 + 4))
print(6.0 + 4, type(6.0 + 4))
print(6 + 4 > 8, type(6 + 4 > 8))
```

- Comparisons can be chained.
- `10 < value < 20` tests both boundaries without repeating `value`.

```python
value = 14
print(10 < value < 20)
print(10 < value and value < 20)
```

- Strings compare lexicographically by Unicode code point.
- Uppercase English letters have smaller code points than lowercase English letters.

```python
print("A" > "a")
print(ord("A"), ord("a"))
```

- Equality works across compatible numeric types.

```python
print(5 == 5.0)
```

- Ordering unrelated types, such as an `int` and a `str`, raises `TypeError`.

```python
print(5 < "6")
```

# 3. Random numbers

- The standard-library `random` module generates pseudorandom values.
- An `import` makes its functions available through the module name.

```python
import random

print(random.random())
```

The three chapter functions differ at their upper boundary.

| Call | Type | Possible values |
|---|---|---|
| `random.random()` | `float` | `0.0 <= value < 1.0` |
| `random.randint(a, b)` | `int` | `a <= value <= b` |
| `random.randrange(a, b)` | `int` | `a <= value < b` |

```python
print(random.random())
print(random.randint(1, 6))
print(random.randrange(1, 6))
```

- Multiplying `random.random()` scales its half-open interval.
- Multiplying by 10 produces a `float` from 0 up to, but not including, 10.

```python
value = random.random() * 10
print(value, 0 <= value < 10)
```

- `randint` includes both endpoints.
- It is the direct choice for an ordinary six-sided die.

```python
die = random.randint(1, 6)
print(die)
```

- `randrange(start, stop, step)` follows the same boundaries as `range`.
- The step restricts which values can appear.

```python
even_number = random.randrange(0, 10, 2)
print(even_number, even_number % 2 == 0)
```

# 4. One-way `if` statements

- A one-way `if` runs its indented block only when the condition is `True`.
- Execution continues after the block either way.

    if condition:
        statements

```python
balance = 125
if balance > 100:
    print("balance is above 100")
print("check complete")
```

- A false condition skips the entire indented block.

```python
balance = 75
if balance > 100:
    print("balance is above 100")
print("check complete")
```

- The colon opens the controlled block.
- Consistent indentation marks every statement inside that block.

```python
credits = 15
if credits >= 12:
    status = "full time"
    print(status)
```

- Separate `if` statements are independent decisions.
- More than one block can run.

```python
number = 30
if number % 5 == 0:
    print("multiple of 5")
if number % 2 == 0:
    print("even")
```

- A comparison already produces the Boolean that a later decision can use.
- No `if` statement is needed just to assign `True` or `False`.

```python
number = 18
is_even = number % 2 == 0
print(is_even)
```

# 5. Two-way `if`/`else` statements

- A two-way selection chooses exactly one branch.
- The `if` block runs for `True`; the `else` block runs for `False`.

    if condition:
        true_branch
    else:
        false_branch

```python
exam_score = 84
if exam_score >= 70:
    print("passing")
else:
    print("not passing")
```

- `else` has no condition.
- It means every case not accepted by the preceding `if`.

```python
seats_open = 0
if seats_open > 0:
    message = "registration available"
else:
    message = "section full"
print(message)
```

- Only the selected branch runs.
- Variables assigned in both branches are available after the statement.

```python
temperature = 76
if temperature >= 80:
    advice = "bring water"
else:
    advice = "regular conditions"
print(advice)
```

- Selection can put two random values in a required order.
- The swap runs only when the first value is smaller.

```python
left = random.randint(0, 9)
right = random.randint(0, 9)
if left < right:
    left, right = right, left
print(left, "-", right, "=", left - right)
```

# 6. Multi-way `if`/`elif`/`else` statements

- A multi-way selection tests conditions from top to bottom.
- The first true branch runs, then the entire chain ends.
- The final `else` handles everything that did not match earlier.

    if condition_1:
        branch_1
    elif condition_2:
        branch_2
    else:
        default_branch

```python
score = 83
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "below B"
print(grade)
```

- Overlapping thresholds belong from most restrictive to least restrictive.
- Once `95 >= 90` succeeds, Python never tests the `>= 80` branch.

```python
score = 95
if score >= 90:
    band = "90 or above"
elif score >= 80:
    band = "80 to 89"
print(band)
```

- Independent `if` statements can report several true facts.
- Here both messages are correct for the same value.

```python
attendance = 100
if attendance >= 100:
    print("perfect")
if attendance >= 60:
    print("meets minimum")
```

- An `elif` chain represents mutually exclusive alternatives.
- Here the same value receives only its highest matching label.

```python
attendance = 100
if attendance >= 100:
    print("perfect")
elif attendance >= 60:
    print("meets minimum")
else:
    print("below minimum")
```

- The final `else` is optional.
- Without it, no branch runs when every condition is false.

```python
score = 65
if score >= 90:
    print("honors")
print("record processed")
```

- A progressive bracket includes tax from completed lower brackets.
- These demonstration rates do not represent any tax authority.

```python
taxable_income = 42_000
if taxable_income <= 12_000:
    tax = taxable_income * 0.10
elif taxable_income <= 40_000:
    tax = 1_200 + (taxable_income - 12_000) * 0.15
else:
    tax = 5_400 + (taxable_income - 40_000) * 0.20
print(f"${tax:,.2f}")
```

# 7. Nested selection

- A nested `if` is an `if` statement inside another branch.
- Python reaches the inner decision only after selecting the outer branch.

```python
prerequisite_met, seats_available = True, False
if prerequisite_met:
    if seats_available:
        registration = "enrolled"
    else:
        registration = "waitlisted"
else:
    registration = "prerequisite required"
print(registration)
```

- Nesting is useful when the second test would be invalid before the first succeeds.
- Division happens only after the denominator is known to be nonzero.

```python
numerator, denominator = 12, 0
if denominator != 0:
    if numerator / denominator > 1:
        print("ratio above 1")
else:
    print("ratio undefined")
```

- Trace nested selection from the outside inward.
- After one branch runs, continue after its matching statement.

```python
x, y = 5, 10
if x < 6:
    if y > 5:
        result = "Product: " + str(x * y)
    else:
        result = "Difference: " + str(y - x)
else:
    result = "x is: " + str(x)
print(result)
```

# 8. Logical operators

Logical operators combine or reverse Boolean values.

| Operator | Result |
|---|---|
| `not p` | the opposite of `p` |
| `p and q` | `True` only when both are `True` |
| `p or q` | `True` when at least one is `True` |

```python
print(not True)
print(True and False)
print(True or False)
```

The complete truth tables show every Boolean input pair.

| `p` | `q` | `p and q` | `p or q` |
|---|---|---|---|
| `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `True` |
| `True` | `False` | `False` | `True` |
| `True` | `True` | `True` | `True` |

```python
p, q = True, False
print(p and q)
print(p or q)
print(not p)
```

- Python combines Boolean expressions with the lowercase operators `and`, `or`, and `not`.

```python
is_python_fun = True
is_java_fun = True
print(is_java_fun or is_python_fun)
print(not is_python_fun)
```

- `and` expresses requirements that must all hold.
- This condition accepts a score only when it is also marked complete.

```python
score = 88
is_complete = True
print(score >= 70 and is_complete)
```

- `or` expresses alternatives.
- This condition accepts either of two valid prerequisite courses.

```python
completed_math = False
completed_statistics = True
print(completed_math or completed_statistics)
```

- Remainders and logical operators express several divisibility rules together.

```python
number = 30
divisible_by_all = number % 2 == 0 and number % 3 == 0 and number % 5 == 0
print(divisible_by_all)
```

- A leap year is divisible by 4 but not by 100, unless it is divisible by 400.
- Parentheses expose the two accepted paths.

```python
year = 2024
is_leap_year = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(is_leap_year)
```

**Example problem.** Compare a two-digit entry with a winning code, then identify an exact, reversed, or shared-digit match.

```python
winning_code, entered_code = 42, 24
win_tens, win_ones = winning_code // 10, winning_code % 10
entry_tens, entry_ones = entered_code // 10, entered_code % 10
exact = entered_code == winning_code
reversed_order = entry_tens == win_ones and entry_ones == win_tens
shared_digit = (entry_tens == win_tens or entry_tens == win_ones
                or entry_ones == win_tens or entry_ones == win_ones)
print(exact, reversed_order, shared_digit)
```

- These conditions overlap, so their order sets the result priority.
- This example checks exact match, reversed match, shared digit, then no match.

```python
if exact:
    drawing_result = "exact match"
elif reversed_order:
    drawing_result = "reversed match"
elif shared_digit:
    drawing_result = "one digit matches"
else:
    drawing_result = "no match"
print(drawing_result)
```

# 9. Precedence and associativity

Python evaluates the operators in a mixed expression by precedence.

| Order | Operators |
|---|---|
| 1 | parentheses `()` |
| 2 | exponentiation `**` |
| 3 | unary `+`, unary `-` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<`, `<=`, `>`, `>=`, `==`, `!=` |
| 7 | `not` |
| 8 | `and` |
| 9 | `or` |

```python
print(3 + 4 * 4 > 5 * (4 + 3) - 1)
print(not 5 < 6)
print(True and False or True)
```

- Arithmetic runs before comparison.
- Comparison runs before `not`, then `and`, then `or`.

```python
result = 2 * 2 - 3 > 2 and 5 >= 6
print(result)
```

- Parentheses override precedence.
- They also make the intended grouping visible when the default order is correct.

```python
print(not (5 < 6))
print((5 < 6) and (7 < 8))
```

- Most arithmetic operators with equal precedence group from left to right.
- Exponentiation groups from right to left.

```python
print(20 / 5 * 2)
print(2 ** 3 ** 2)
print((2 ** 3) ** 2)
```

- Comparisons chain instead of grouping into a Boolean intermediate.
- `a < b < c` means `a < b and b < c`.

```python
a, b, c = 10, 15, 20
print(a < b < c)
print(a < b and b < c)
```

- Chained assignment evaluates the right side once.
- The resulting value is then bound to every name on the left.

```python
a = b = c = 5
print(a, b, c)
```

# 10. Short-circuit evaluation

- `and` stops as soon as one operand is false.
- A later operand cannot change the result to `True`.

```python
print(False and (10 / 0 > 1))
```

- `or` stops as soon as one operand is true.
- A later operand cannot change the result to `False`.

```python
print(True or (10 / 0 > 1))
```

- Put a safety check before the expression it protects.
- The division is reached only when the denominator is nonzero.

```python
numerator, denominator = 12, 0
ratio_is_large = denominator != 0 and numerator / denominator > 1
print(ratio_is_large)
```

- Put inexpensive checks before expensive work when either order has the same meaning.
- Short-circuiting can then avoid the later work.

```python
has_submission = False
score = 92
eligible = has_submission and score >= 70
print(eligible)
```

- With Boolean operands, `and` and `or` produce Boolean results.
- Other operand types use truthiness and return one selected operand.

```python
print("ready" or "backup")
print("" or "backup")
print(bool("ready" or "backup"))
```

# 11. Range and location tests

- A chained comparison is the direct Python spelling for a bounded range.
- The operators decide whether each endpoint is included.

```python
value = 15
print(10 < value < 20)
print(10 <= value <= 20)
```

- Reverse the whole range test with `not` to detect values outside it.

```python
value = 25
outside = not 10 <= value <= 20
print(outside)
```

- `in` tests membership in a collection.
- It is clearer than repeating equality with `or`.

```python
day = "Saturday"
print(day in ("Saturday", "Sunday"))
print(day == "Saturday" or day == "Sunday")
```

- A point is inside a circle when its squared distance from the center is less than the
  squared radius.
- Comparing the squares avoids an unnecessary square root.

    (point_x - center_x) ** 2 + (point_y - center_y) ** 2 < radius ** 2

```python
center_x, center_y, radius = 2, 3, 5
point_x, point_y = 5, 7
distance_squared = (point_x - center_x) ** 2 + (point_y - center_y) ** 2
print(distance_squared < radius ** 2)
```

- `<` and `>` distinguish the two sides of the boundary.
- `math.isclose` recognizes a computed floating-point value on the boundary.

```python
import math
if math.isclose(distance_squared, radius ** 2):
    location = "on"
elif distance_squared < radius ** 2:
    location = "inside"
else:
    location = "outside"
print(location)
```

# 12. Conditional expressions

- A conditional expression chooses one of two expression values.
- Its order is value for true, condition, value for false.

    value_if_true if condition else value_if_false

```python
x = 8
y = 1 if x > 0 else -1
print(y)
```

- A short `if`/`else` assignment can become one conditional expression.
- Both forms bind the same result.

```python
number = 17
label = "even" if number % 2 == 0 else "odd"
print(label)
```

- Only the selected expression is evaluated.
- The unselected expression cannot raise an error or cause a side effect.

```python
safe_value = 1 if True else 10 / 0
print(safe_value)
```

- Conditional expressions fit short, symmetric choices.
- Use a full statement when either branch needs several actions.

```python
score = 68
message = "pass" if score >= 70 else "review and retry"
print(message)
```

# 13. `match`/`case` statements

- `match`/`case` selects a branch by pattern.
- It requires Python 3.10 or later.

```python
import sys

print(sys.version_info >= (3, 10))
```

- The subject after `match` is evaluated once.
- The first matching `case` runs.
- `_` is the optional default pattern.

```python
menu_choice = 2
match menu_choice:
    case 1:
        print("open")
    case 2:
        print("save")
    case _:
        print("unknown choice")
```

- Literal patterns can be numbers, strings, `True`, `False`, or `None`.
- Python leaves the `match` after one case, so no `break` is needed.

```python
command = "write"
match command:
    case "save" | "write":
        action = "store changes"
    case "quit" | "exit":
        action = "close program"
    case _:
        action = "show help"
print(action)
```

- Boolean literals are patterns too.
- They match the Boolean values directly.

```python
is_registered = True
match is_registered:
    case True:
        print("registered")
    case False:
        print("not registered")
```

- A guard adds a Boolean condition after a pattern.
- `case value if condition` runs only when both the pattern and guard match.

```python
score = 87
match score:
    case value if value >= 90:
        band = "A"
    case value if value >= 80:
        band = "B"
    case _:
        band = "below B"
print(band)
```

- A bare name in a `case` captures the subject instead of comparing with an existing
  variable.
- Use literal patterns for exact values and guards for conditions.

```python
status = 7
match status:
    case 1:
        print("one")
    case captured:
        print("captured", captured)
```

- Use `if`/`elif` for ordered ranges.
- Use `match` when exact shapes or values make the alternatives clearer.

```python
score = 87
if score >= 80:
    print("range test")
match score:
    case 87:
        print("exact value")
```

# 14. Common selection errors

- Parentheses around a condition are optional.
- The colon and indentation are required.
- The commented lines mark the two common mistakes.

```python
score = 88
if score >= 70:
    print("passing")
# if score >= 70
# print("passing")
```

- `=` assigns; `==` compares.
- An assignment cannot replace the comparison in an `if` condition.

```python
score = 90
# if score = 90:
#     print("exact score")
print(score == 90)
```

- `else` never has a condition.
- A second condition belongs after `elif`.

```python
score = 65
if score >= 70:
    print("passing")
elif score < 70:
    print("not passing")
```

- Indentation controls which statements belong to a branch.
- An unindented statement runs after the selection.

```python
is_open = False
if is_open:
    message = "come in"
else:
    message = "closed"
print(message)
```

- Python has no increment operator.
- `++count` applies unary plus twice and does not change the value.
- Use `count += 1`.

```python
count = 5
++count
print(count)
count += 1
print(count)
```

# 15. Program: body mass index category

**Requirement.** Given an adult's mass and positive height, calculate body mass index
(BMI) and report its standard category.

| | |
|---|---|
| Input | mass in kilograms and height in meters |
| Process | divide mass by height squared, then select a category by threshold |
| Output | BMI to one decimal place and its category |

The calculation is

    bmi = mass_kg / height_m ** 2

```python
# body_mass_kg = float(input("Mass in kilograms: "))
# height_m = float(input("Height in meters: "))
body_mass_kg = 72.5
height_m = 1.75
```

**Process.** Test thresholds from lowest to highest so the first true branch identifies
the one category containing the value.

```python
bmi = body_mass_kg / height_m ** 2
if bmi < 18.5:
    bmi_category = "underweight"
elif bmi < 25:
    bmi_category = "healthy range"
elif bmi < 30:
    bmi_category = "overweight"
else:
    bmi_category = "obesity range"
```

**Output.** Keep the numeric result and category together.

```python
print(f"BMI: {bmi:.1f}")
print("Category:", bmi_category)
```

# 16. Program: point location

**Requirement.** Given a circle with a positive radius and a point, report whether the
point is inside, on, or outside the circle.

| | |
|---|---|
| Input | center coordinates, radius, and point coordinates |
| Process | compare squared distance from the center with squared radius |
| Output | the point's location relative to the circle |

The squared-distance comparison is

    (point_x - center_x) ** 2 + (point_y - center_y) ** 2  ?  radius ** 2

```python
# center_x, center_y = map(float, input("Circle center x y: ").split())
# radius = float(input("Radius: "))
# point_x, point_y = map(float, input("Point x y: ").split())
center_x, center_y = 2.0, 3.0
radius = 5.0
point_x, point_y = 5.0, 7.0
```

**Process.** Calculate each side once, then use a tolerant comparison at the floating-point
boundary.

```python
import math
dx, dy = point_x - center_x, point_y - center_y
distance_squared, radius_squared = dx ** 2 + dy ** 2, radius ** 2
if math.isclose(distance_squared, radius_squared):
    point_location = "on"
elif distance_squared < radius_squared:
    point_location = "inside"
else:
    point_location = "outside"
```

**Output.** Name the point and its location.

```python
print(f"Point ({point_x}, {point_y}) is {point_location} the circle.")
```
