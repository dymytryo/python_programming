# 1. Running code and displaying results

A program is a sequence of statements the interpreter executes in order, top to bottom.
A statement is one instruction. An expression is anything that produces a value, such as
`2 + 3`.

A notebook cell displays two separate things:

- everything that `print` writes while the cell runs, and
- the value of the last expression in the cell, if that expression has a value.

The second one is a notebook convenience. In a program saved as a `.py` file, nothing is
displayed unless `print` says so.

```python
2 + 3
```

That `5` came from the last-expression rule, not from `print`. Note the `Out[ ]` label

Only the last expression is displayed, so intermediate values in a cell are lost unless

# 2. Numbers and arithmetic

You can compute with Python before knowing anything about variables. Two numeric types
carry every value in this notebook.

| Type | Written as | Holds |
|---|---|---|
| `int` | `3`, `-8`, `1_000_000` | whole numbers, no fractional part |
| `float` | `3.0`, `-0.5`, `2.5e3` | numbers with a fractional part |

`type(value)` reports which one you have. Underscores inside a numeric literal are digit
separators and the interpreter ignores them, so `1_000_000` and `1000000` are the same
number.

`int` has no fixed width. It grows to as many digits as memory allows, so integer
arithmetic never wraps around and never loses precision.

```python
print(type(3), type(3.0))
print(1_000_000 == 1000000)

# An int grows to fit. This is a 61-digit number, exact.
print(2 ** 200)
```

## The seven arithmetic operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `34 + 1` | `35` |
| `-` | subtraction | `34.0 - 0.1` | `33.9` |
| `*` | multiplication | `300 * 30` | `9000` |
| `/` | float division | `1 / 2` | `0.5` |
| `//` | floor division | `1 // 2` | `0` |
| `%` | remainder | `20 % 3` | `2` |
| `**` | exponentiation | `4 ** 0.5` | `2.0` |

Mixing an `int` and a `float` promotes the result to `float`, and `/` always produces a
`float` even when the division comes out even.

```python
print("34 + 1     =", 34 + 1)
print("34.0 - 0.1  =", 34.0 - 0.1)
print("300 * 30    =", 300 * 30)
print("1 / 2       =", 1 / 2)
print("4 / 2       =", 4 / 2)      # even division, still a float
print("7 + 0.5     =", 7 + 0.5, type(7 + 0.5))
```

## Two kinds of quotient

`/` gives the true quotient as a `float`. `//` gives the floor of that quotient: the
largest whole number less than or equal to it.

With positive operands, floor division looks like chopping off the decimals. With a
negative operand it does not, because it floors downward rather than toward zero:
`-7 // 2` is `-4`, not `-3`. The result type follows the operands, so two ints give an
int and any float gives a float.

`**` accepts a fractional exponent, which makes `4 ** 0.5` the square root.

```python
print("7 / 2    =", 7 / 2)
print("7 // 2   =", 7 // 2)       # floor: 3
print("-7 // 2  =", -7 // 2)      # floor: -4, not -3
print("7.0 // 2 =", 7.0 // 2)     # a float operand gives a float
print("4 ** 0.5 =", 4 ** 0.5)     # square root
print("2 ** 10  =", 2 ** 10)
```

## Precedence

Python evaluates an expression by the ordinary arithmetic rules.

| Order | Operators | Grouping |
|---|---|---|
| 1 | `()` | innermost pair first |
| 2 | `**` | right to left |
| 3 | unary `-` | |
| 4 | `*` `/` `//` `%` | left to right |
| 5 | `+` `-` | left to right |

Two rows cause most wrong answers. `**` binds tighter than everything else, so
`2 + 2 ** 3 / 2` is `2 + 8 / 2`, which is `6.0`. And `**` groups right to left, so
`2 ** 3 ** 2` is `2 ** 9`, not `8 ** 2`. Where the intended order is not obvious at a
glance, add the parentheses even when they are redundant.

```python
print("2 + 2 ** 3 / 2 =", 2 + 2 ** 3 / 2)
print("2 ** 3 ** 2    =", 2 ** 3 ** 2)     # right to left: 2 ** 9
print("(2 ** 3) ** 2  =", (2 ** 3) ** 2)
print("-2 ** 2        =", -2 ** 2)         # ** binds tighter than unary minus
print("(-2) ** 2      =", (-2) ** 2)
```

## Translating algebra into Python

Mathematical notation leaves two things implicit that Python requires you to write out:
every multiplication needs a `*`, and every fraction bar becomes a `/` with parentheses
around whatever was above and below it.

    3 + 4x
    ------ + 10(y - 5)     becomes     (3 + 4 * x) / 5 + 10 * (y - 5)
       5

Drop the parentheses around `3 + 4 * x` and you have computed `3 + (4x/5)` instead, which
is a different number and raises no error.

```python
# With x = 2 and y = 7:
print("correct:            ", (3 + 4 * 2) / 5 + 10 * (7 - 5))
print("missing parentheses:", 3 + 4 * 2 / 5 + 10 * (7 - 5))
```

# 3. Variables and names

A variable is a name bound to a value. The assignment statement is

    name = expression

and it runs in that order: evaluate the expression on the right, then bind the name on
the left to the result. Read `=` as "becomes", never as "equals".

That order is the whole explanation of `x = x + 1`. Python computes the current value of
`x` plus one, and only then rebinds `x` to the new value.

```python
enrolled = 24
print("enrolled:", enrolled)

enrolled = enrolled + 3     # right side evaluates to 27, then rebind
print("enrolled:", enrolled)
```

Three consequences.

A name must be assigned before it is used. Using an unassigned name is a `NameError`, and
it is the error you will see most often, usually from a typo in the name.

Assigning again replaces the binding, and the previous value is gone.

The name carries no type of its own. The type comes from the value, so the same name can
hold an `int` on one line and a `float` on the next.

```python
points = 3
print(points, type(points))

points = 3.5                 # same name, different type
print(points, type(points))

first = second = third = 0   # chained assignment: one value, three names
print(first, second, third)
```

## Identifiers

An identifier is the name you give a variable, a function, or a module. The rules:

- Letters, digits, and underscores (`_`) only. No spaces, no hyphens, no operators. `*`
  is multiplication and is never part of a name.
- It cannot start with a digit. `score1` is legal, `1score` is not.
- It cannot be a reserved word. Reserved words, also called keywords, already have a
  meaning in the language.
- Any length, and case sensitive: `total`, `Total`, and `TOTAL` are three different
  names.

The conventions layered on top of the rules matter as much as the rules themselves:
`lower_snake_case` for variables, `UPPER_SNAKE_CASE` for constants, and a name that
describes the value. `credit_hours` is worth more than `ch` every time someone reads the
program, including you in a week.

```python
import keyword

print(len(keyword.kwlist), "reserved words, none of them available as a name:")
print(keyword.kwlist)

# Case sensitivity: three separate variables, not one.
total = 10
Total = 20
TOTAL = 30
print(total, Total, TOTAL)
```

# 4. Text and numbers are different things

A string, type `str`, is text: any characters between a matching pair of quotes. Single
and double quotes both work, and choosing the other one is how you include a quote in the
text.

The arithmetic operators exist for strings too, and they do something else entirely:
`+` joins two strings end to end, `*` repeats a string a whole number of times. Neither
one looks at the text as a number, so `"8" + "5"` is `"85"`, not `13`.

Mixing the two types raises `TypeError`, because Python refuses to guess whether you
meant to add or to join. Convert first, with `int()`, `float()`, or `str()`.

```python
course = "BCIS 1305"
section = '004'

print(course + " section " + section)   # + joins
print("-" * 30)                         # * repeats
print("8" + "5")                         # two strings, joined
print(8 + 5)                             # two numbers, added
```

```python
score_text = "88"

# Convert, then compute.
print(int(score_text) + 2)

# Convert the other way to join a number onto text.
print("Score: " + str(88))

# f-strings avoid the conversion entirely: put the value in braces.
print(f"Score: {88}, doubled: {88 * 2}")
```

```python
# Mixing them without converting is a TypeError.
try:
    print("Score: " + 88)
except TypeError as error:
    print("TypeError:", error)
```

# 7. Floats in the real world

A `float` stores an approximation of a decimal value in a fixed 64 bits, in binary. Most
decimal fractions have no exact binary form, so small representation errors appear in
results. `0.1 + 0.2` is not `0.3`.

This is a property of the storage format used by the hardware, not a Python defect. Every
language using the same format behaves the same way. The practical rules that follow from
it: never test two computed floats for exact equality, and never let a raw float be the
thing a user reads.

```python
print("0.1 + 0.2        =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3  ->", 0.1 + 0.2 == 0.3)

# Money shows it immediately.
print("189.00 * 0.0825   =", 189.00 * 0.0825)
```

## Converting and rounding

`int(value)` truncates toward zero. It drops the fractional part rather than rounding, so
`int(4.9)` is `4` and `int(-4.9)` is `-4`. Use it when the whole part is what you want,
such as counting complete weeks in a term.

`round(value)` rounds to the nearest whole number and returns an `int`. On an exact tie it
rounds to the even neighbour, so `round(0.5)` is `0` and `round(1.5)` is `2`. This is
deliberate: always rounding ties up would bias a long column of numbers upward.

`round(value, n)` keeps `n` digits after the decimal point and returns a `float`. It can
still surprise you, because the value handed to it is already a binary approximation:
`round(2.675, 2)` gives `2.67`, since the stored value sits just below 2.675.

For anything a person reads, format instead of rounding. The f-string format
`f"{value:.2f}"` always produces exactly two decimal places, and it leaves the underlying
number alone.

```python
print("int(4.9)         =", int(4.9))       # truncates toward zero
print("int(-4.9)        =", int(-4.9))
print("round(4.9)       =", round(4.9))
print("round(0.5)       =", round(0.5), " round(1.5) =", round(1.5))
print("round(2.675, 2)  =", round(2.675, 2))
```

```python
purchase = 189.00
tax = purchase * 0.0825
total = purchase + tax

print("raw:      ", total)
print("formatted:", f"${total:.2f}")
print("aligned:  ", f"${total:>10.2f}")   # width 10, right aligned
```

# 8. Integer division and remainder in practice

`%` returns what is left after floor division. For any `a` and any non-zero `b`:

    a == (a // b) * b + (a % b)

Those two operators together do three jobs you will use constantly.

**Testing divisibility.** `n % 2` is `0` for an even number and `1` for an odd one.

**Splitting a total into units.** `//` gives the whole units, `%` gives the remainder.
Total minutes into hours and minutes, total seconds into minutes and seconds, cents into
dollars and cents.

**Wrapping around a cycle.** With the seven weekdays numbered 0 to 6, the day 10 days
after day 6 is `(6 + 10) % 7`, which is day 2. The remainder keeps a growing count inside
a fixed range.

One rule to remember: the result takes the sign of the right operand, so `-7 % 3` is `2`,
not `-1`.

```python
class_size = 23
print("can the class pair up?", class_size % 2 == 0)

total_minutes = 275
print(total_minutes, "minutes =", total_minutes // 60, "h", total_minutes % 60, "min")

SATURDAY = 6
print("10 days after day 6 is day", (SATURDAY + 10) % 7)

print("-7 % 3 =", -7 % 3)
print("identity holds:", (-7 // 3) * 3 + (-7 % 3))
```

# 9. Assignment shortcuts

## Augmented assignment

An augmented assignment reads the variable, applies the operator, and rebinds the same
name. Every arithmetic operator has one.

| Operator | Example | Equivalent |
|---|---|---|
| `+=` | `i += 8` | `i = i + 8` |
| `-=` | `i -= 8` | `i = i - 8` |
| `*=` | `i *= 8` | `i = i * 8` |
| `/=` | `i /= 8` | `i = i / 8` |
| `//=` | `i //= 8` | `i = i // 8` |
| `%=` | `i %= 8` | `i = i % 8` |
| `**=` | `i **= 8` | `i = i ** 8` |

Because it reads the variable first, the variable must already hold a value: `count += 1`
on an unassigned `count` is a `NameError`.

The operator applies to the whole expression on the right, not only to its first term.
`i *= j + 5` means `i = i * (j + 5)`, not `i = i * j + 5`.

```python
graded = 0
to_grade = 30

graded += 5        # graded = graded + 5
to_grade -= 5      # to_grade = to_grade - 5
print("graded:", graded, " still to grade:", to_grade)

i, j = 2, 3
i *= j + 5         # i = i * (3 + 5) = 16, not i * 3 + 5 = 11
print("i:", i)
```

## Simultaneous assignment

    var1, var2, ..., varn = exp1, exp2, ..., expn

Every expression on the right is evaluated before any name on the left is bound. The right
side is packed into a tuple, then unpacked into the names from left to right.

That is why swapping two variables needs no temporary: `y, x` is built from the old
values, so `x, y = y, x` cannot overwrite a value it still needs. The counts on the two
sides must match, otherwise Python raises `ValueError`.

```python
midterm, final = 82, 91
print("before:", midterm, final)

midterm, final = final, midterm      # no temporary variable needed
print("after: ", midterm, final)

try:
    a, b, c = 1, 2                   # three names, two values
except ValueError as error:
    print("ValueError:", error)
```

# 11. Very large and very small numbers

## Scientific notation

Scientific notation writes a float as a mantissa times a power of ten. Python writes the
power with `e` or `E`, in either case.

| Literal | Means | Value |
|---|---|---|
| `1.23456e2` | 1.23456 x 10^2 | 123.456 |
| `1.23456e+2` | the same; the `+` is optional | 123.456 |
| `1.23456e-2` | 1.23456 x 10^-2 | 0.0123456 |

A literal written this way is a `float` even when the value is whole, so `2e3` is
`2000.0`. Python also switches to this form on output once the decimal form would be
unreadable.

## Overflow and underflow

Overflow is a result too large for the type that must hold it. `int` has no fixed size, so
integers never overflow. `float` is fixed at 64 bits with a largest value near
1.8 x 10^308, so a float computation past that limit raises `OverflowError`, while
arithmetic on values that are already floats gives `inf`, the floating-point infinity,
instead of raising.

Underflow is the other end: a float too close to zero to represent. Python approximates it
to `0.0` and says nothing, which is usually harmless. The one thing to watch is dividing by
a value that has underflowed, since that is a `ZeroDivisionError`.

```python
print(1.23456e2, 1.23456e+2, 1.23456e-2)
print("2e3 is", 2e3, type(2e3))

print(0.0001)     # still shown as a decimal
print(0.00001)    # switches to e-notation
```

```python
import sys

print("digits in 245 ** 1000:", len(str(245 ** 1000)))   # int: exact, no limit
print("largest float:", sys.float_info.max)

try:
    245.0 ** 1000                    # float: past the limit
except OverflowError as error:
    print("OverflowError:", error)

print("1e308 * 10    =", 1e308 * 10)      # already a float, so inf
print("1e-320 / 1e10 =", 1e-320 / 1e10)   # underflow, approximated to 0.0
```

# 12. Reading the clock

`time.time()` returns the number of seconds elapsed since midnight, 1 January 1970,
Coordinated Universal Time (UTC). That reference point is called the epoch, and it dates
from the introduction of the Unix operating system. The value is a float, with fractions of
a second after the decimal point.

An `import` statement makes a module's contents available under its name, and belongs at
the top of the program.

Turning that count into a clock time is nothing but the two operators from part 8:

    total seconds since the epoch
        current second   = total seconds % 60
        total minutes    = total seconds // 60
        current minute   = total minutes % 60
        total hours      = total minutes // 60
        current UTC hour = total hours % 24

```python
import time

current_time = time.time()
print("seconds since the epoch:", current_time)

total_seconds = int(current_time)
current_second = total_seconds % 60
total_minutes = total_seconds // 60
current_minute = total_minutes % 60
total_hours = total_minutes // 60
current_hour = total_hours % 24

# :02d pads a single digit with a leading zero, so 16:4:9 reads as 16:04:09.
print(f"{current_hour:02d}:{current_minute:02d}:{current_second:02d} UTC")
```

Because the value is a float measured in seconds, the same function is a stopwatch: read it
before and after a computation, then subtract.

```python
import time

import math

start = time.time()
value = math.factorial(200_000)         # big enough to take measurable time
elapsed = time.time() - start

print("bits needed to store 200000!:", value.bit_length())
print(f"elapsed: {elapsed:.4f} seconds")
```

# 13. From problem to program

The software development process is the ordered set of phases that turns a problem
statement into working, maintained software. The phases are the same whether the result is
a 20-line class exercise or a system with a thousand users. Only the amount of work in each
phase changes.

    requirement specification
              |
              v
        system analysis          input -> process -> output
              |
              v
         system design
              |
              v
        implementation
              |
              v
           testing
              |
              v
         deployment  ->  maintenance  ->  back to requirements

| Phase | What it produces |
|---|---|
| Requirement specification | A written statement of what the software must do, agreed with the people who will use it. |
| System analysis | The data flow: what goes in, what comes out, what the system has to know. |
| System design | The components, what each is responsible for, and how they relate. |
| Implementation | Code for each component, written, debugged, and working together. |
| Testing | Evidence that the code meets the requirement specification, ideally gathered by someone who did not write it. |
| Deployment | The software made available to its users. |
| Maintenance | Fixes and changes as the requirements and the environment move. |

The core of analysis and design is IPO: input, process, output. Name the inputs, name the
outputs, then work out the process that connects them. On a program this small it takes two
minutes on paper, and it is what stops you from writing the program three times.

## Tracing

Tracing is stepping through the statements by hand and writing down what each variable
holds after each one. It is how you find a wrong answer that raises no error, and it is the
skill that makes the two programs below debuggable.

| Statement | `credit_hours` | `cost_per_hour` | `tuition` |
|---|---|---|---|
| `credit_hours = 12` | 12 | does not exist | does not exist |
| `cost_per_hour = 275.50` | 12 | 275.50 | does not exist |
| `tuition = credit_hours * cost_per_hour` | 12 | 275.50 | 3306.0 |

A variable does not exist until a statement assigns to it, each assignment replaces the
previous value, and nothing appears on screen until `print` runs.

```python
# The same three statements, printing the state after each one.
credit_hours = 12
print("after 1: credit_hours  =", credit_hours)

cost_per_hour = 275.50
print("after 2: cost_per_hour =", cost_per_hour)

tuition = credit_hours * cost_per_hour
print("after 3: tuition       =", tuition)
```

# 14. Program: monthly loan payment

**Requirement.** Given a loan amount, an annual interest rate, and a term in years, report
the monthly payment and the total amount repaid.

**IPO.**

| | |
|---|---|
| Input | annual interest rate as a percent, term in years, loan amount |
| Process | convert the annual percent to a monthly rate, apply the payment formula, multiply by the number of payments |
| Output | monthly payment and total payment, each to two decimals |

The monthly payment on a fixed-rate loan is

    payment = amount * monthly_rate / (1 - 1 / (1 + monthly_rate) ** months)

Both conversions in the process step are where this program goes wrong in practice. An
annual rate of 5.5 percent is `5.5 / 100 / 12` per month, not `5.5 / 12`. A 10-year term is
`10 * 12` payments, not 10. Naming the two constants makes both conversions visible on the
line where they happen.

```python
MONTHS_PER_YEAR = 12
PERCENT = 100

# Live version:
#   annual_interest_rate = float(input("Annual interest rate in percent: "))
#   years = int(input("Term in years: "))
#   loan_amount = float(input("Loan amount: "))
annual_interest_rate = 5.5
years = 10
loan_amount = 27000.00

monthly_rate = annual_interest_rate / PERCENT / MONTHS_PER_YEAR
months = years * MONTHS_PER_YEAR

monthly_payment = loan_amount * monthly_rate / (1 - 1 / (1 + monthly_rate) ** months)
total_payment = monthly_payment * months

print(f"Loan amount:     ${loan_amount:>10,.2f}")
print(f"Monthly payment: ${monthly_payment:>10,.2f}")
print(f"Total repaid:    ${total_payment:>10,.2f}")
print(f"Interest paid:   ${total_payment - loan_amount:>10,.2f}")
```

Change `years` to 20 and run it again. The monthly payment falls and the interest paid
roughly doubles, which is the whole argument about loan terms in two numbers.

# 15. Program: distance between two points

**Requirement.** Given two points, report the straight-line distance between them.

**IPO.**

| | |
|---|---|
| Input | `x1`, `y1`, `x2`, `y2` |
| Process | subtract to get the horizontal and vertical differences, then apply the distance formula |
| Output | the distance, to two decimals |

The distance between (x1, y1) and (x2, y2) is

    ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

The exponent `0.5` is the square root, so this needs nothing beyond part 2. The `math`
module spells the same computation two clearer ways: `math.sqrt(value)`, and
`math.hypot(dx, dy)`, which takes the two differences directly and stays accurate on
coordinates large enough to overflow the squares.

```python
import math

x1, y1 = 1.5, -3.4
x2, y2 = 4.0, 5.9

dx = x2 - x1
dy = y2 - y1

print("with operators:", (dx ** 2 + dy ** 2) ** 0.5)
print("with math.sqrt:", math.sqrt(dx ** 2 + dy ** 2))
print("with math.hypot:", math.hypot(dx, dy))

print(f"\nDistance from ({x1}, {y1}) to ({x2}, {y2}) is {math.hypot(dx, dy):.2f}")
```
