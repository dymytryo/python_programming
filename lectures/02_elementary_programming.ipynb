# 2. Numbers and arithmetic

Two numeric types carry every value in this notebook.

| Type | Written as | Holds |
|---|---|---|
| `int` | `3`, `-8`, `1_000_000` | whole numbers, no fractional part |
| `float` | `3.0`, `-0.5`, `2.5e3` | numbers with a fractional part |

`type(value)` reports which one you have.

```python
print(type(3))
print(type(3.0))
```

An `int` has no fixed width. It grows to as many digits as memory allows, so integer
arithmetic never wraps around and never loses precision.

```python
print(2 ** 200)
```

Underscores inside a number are digit separators. The interpreter ignores them.

```python
print(1_000_000 == 1000000)
```

Seven arithmetic operators.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `34 + 1` | `35` |
| `-` | subtraction | `34.0 - 0.1` | `33.9` |
| `*` | multiplication | `300 * 30` | `9000` |
| `/` | float division | `1 / 2` | `0.5` |
| `//` | floor division | `1 // 2` | `0` |
| `%` | remainder | `20 % 3` | `2` |
| `**` | exponentiation | `4 ** 0.5` | `2.0` |

```python
print(34 + 1)
print(34.0 - 0.1)
print(300 * 30)
```

`/` always produces a `float`, even when the division comes out even.

```python
print(1 / 2)
print(4 / 2)
```

`//` produces the floor of the quotient: the largest whole number less than or equal to the
true result.

```python
print(7 / 2)
print(7 // 2)
```

Floor means downward, not toward zero. With a negative operand the two differ.

```python
print(-7 // 2)     # floor: -4
print(int(-7 / 2))  # toward zero: -3
```

The result type follows the operands. Two ints give an int; any float gives a float.

```python
print(7 // 2)
print(7.0 // 2)
```

`%` gives what is left over after floor division.

```python
print(20 % 3)
```

`**` raises to a power, and accepts a fractional exponent, which makes `0.5` the square
root.

```python
print(2 ** 10)
print(4 ** 0.5)
```

Mixing an `int` and a `float` promotes the result to `float`.

```python
print(7 + 0.5, type(7 + 0.5))
```

Python evaluates an expression by the ordinary arithmetic rules.

| Order | Operators | Grouping |
|---|---|---|
| 1 | `()` | innermost pair first |
| 2 | `**` | right to left |
| 3 | unary `-` | |
| 4 | `*` `/` `//` `%` | left to right |
| 5 | `+` `-` | left to right |

`**` binds tighter than everything else, so this is `2 + 8 / 2`.

```python
print(2 + 2 ** 3 / 2)
```

`**` groups right to left, so `2 ** 3 ** 2` is `2 ** 9`, not `8 ** 2`.

```python
print(2 ** 3 ** 2)
print((2 ** 3) ** 2)
```

`**` also binds tighter than a leading minus sign.

```python
print(-2 ** 2)
print((-2) ** 2)
```

Algebraic notation leaves two things implicit that Python makes you write: every
multiplication needs a `*`, and every fraction bar becomes a `/` with parentheses around
what was above and below it.

    3 + 4x
    ------ + 10(y - 5)     becomes     (3 + 4 * x) / 5 + 10 * (y - 5)
       5

Drop those parentheses and you have computed `3 + 4x/5` instead. No error, different
number. With x = 2 and y = 7:

```python
print((3 + 4 * 2) / 5 + 10 * (7 - 5))   # correct
print(3 + 4 * 2 / 5 + 10 * (7 - 5))     # parentheses missing
```

# 3. Variables and names

A variable is a name bound to a value. An assignment statement is `name = expression`, and
it runs right side first: evaluate the expression, then bind the name. Read `=` as
"becomes".

```python
enrolled = 24
print(enrolled)
```

That order is the whole explanation of `x = x + 1`: the current value is read, three is
added, and only then is the name rebound.

```python
enrolled = enrolled + 3
print(enrolled)
```

Assigning again replaces the binding. The previous value is gone.

```python
enrolled = 30
print(enrolled)
```

A name must be assigned before it is used. Using one that was never assigned, usually
because of a typo, is a `NameError`.

```python
try:
    print(enrolled_students)
except NameError as error:
    print("NameError:", error)
```

The name carries no type of its own. The type comes from the value, so the same name can
hold an `int` and then a `float`.

```python
points = 3
print(points, type(points))

points = 3.5
print(points, type(points))
```

One expression can be bound to several names at once.

```python
first = second = third = 0
print(first, second, third)
```

An identifier is the name you give a variable, a function, or a module. Letters, digits and
underscores only, never starting with a digit, and never a reserved word. Any length, and
case sensitive.

```python
import keyword

print(len(keyword.kwlist), "reserved words:")
print(keyword.kwlist)
```

Case sensitivity is not a detail. These are three separate variables.

```python
total = 10
Total = 20
TOTAL = 30
print(total, Total, TOTAL)
```

On top of the rules sit the conventions: `lower_snake_case` for variables,
`UPPER_SNAKE_CASE` for constants, and a name that describes the value. `credit_hours` beats
`ch` every time anyone reads the program, including you next week.

# 4. Text and numbers are different things

A string, type `str`, is text between a matching pair of quotes. Single and double quotes
both work, and having two kinds is how you put a quote inside the text.

```python
course = "BCIS 1305"
section = '004'
print(course, section)
```

`+` joins two strings end to end.

```python
print("BCIS" + " " + "1305")
```

`*` repeats a string a whole number of times.

```python
print("-" * 30)
```

The same two operators on numbers do arithmetic instead. Digits inside quotes are still
text, so `+` joins them.

```python
print("8" + "5")
print(8 + 5)
```

Mixing the two types is a `TypeError`. Python will not guess whether you meant to add or
to join.

```python
try:
    print("Score: " + 88)
except TypeError as error:
    print("TypeError:", error)
```

`int()` and `float()` turn text into a number.

```python
score_text = "88"
print(int(score_text) + 2)
```

`str()` goes the other way, so a number can be joined onto text.

```python
print("Score: " + str(88))
```

An f-string skips the conversion: put an `f` before the quote and the value in braces.

```python
score = 88
print(f"Score: {score}, doubled: {score * 2}")
```

# 7. Floats in the real world

A `float` stores an approximation of a decimal value in a fixed 64 bits, in binary. Most
decimal fractions have no exact binary form, so small representation errors appear in
results.

```python
print(0.1 + 0.2)
```

Which means computed floats must never be compared for exact equality.

```python
print(0.1 + 0.2 == 0.3)
```

Money shows it on the first multiplication.

```python
print(189.00 * 0.0825)
```

`int()` truncates toward zero. It drops the fractional part rather than rounding.

```python
print(int(4.9))
print(int(-4.9))
```

`round()` rounds to the nearest whole number and returns an `int`.

```python
print(round(4.9))
print(round(4.4))
```

On an exact tie it rounds to the even neighbour. Always rounding ties up would bias a long
column of numbers upward.

```python
print(round(0.5))
print(round(1.5))
print(round(2.5))
```

`round(value, n)` keeps `n` digits after the decimal point and returns a `float`.

```python
print(round(15.5925, 2))
```

It can still look wrong, because the value handed to it is already a binary approximation.
The stored form of 2.675 sits just below 2.675.

```python
print(round(2.675, 2))
```

So for anything a person reads, format instead of rounding. `f"{value:.2f}"` always gives
exactly two decimal places and leaves the number alone.

```python
total = 189.00 * 1.0825
print(total)
print(f"${total:.2f}")
```

The same format field sets a width and adds thousands separators, which is how columns of
money line up.

```python
print(f"${total:>12,.2f}")
print(f"${total * 100:>12,.2f}")
```

# 8. Integer division and remainder in practice

`//` and `%` are two halves of one division. For any `a` and any non-zero `b`:

    a == (a // b) * b + (a % b)

```python
a, b = 275, 60
print((a // b) * b + (a % b))
```

**Job one: testing divisibility.** `n % 2` is `0` for an even number and `1` for an odd
one.

```python
class_size = 23
print(class_size % 2)
print(class_size % 2 == 0)
```

**Job two: splitting a total into units.** `//` gives the whole units, `%` gives what is

```python
total_minutes = 275
print(total_minutes // 60, "hours", total_minutes % 60, "minutes")
```

**Job three: wrapping around a cycle.** With the weekdays numbered 0 to 6, `% 7` keeps a

```python
SATURDAY = 6
print((SATURDAY + 10) % 7)
```

One rule to carry: the remainder takes the sign of the right operand.

```python
print(-7 % 3)
print(7 % -3)
```

# 9. Assignment shortcuts

An augmented assignment reads the variable, applies the operator, and rebinds the same name.

| Operator | Example | Equivalent |
|---|---|---|
| `+=` | `i += 8` | `i = i + 8` |
| `-=` | `i -= 8` | `i = i - 8` |
| `*=` | `i *= 8` | `i = i * 8` |
| `/=` | `i /= 8` | `i = i / 8` |
| `//=` | `i //= 8` | `i = i // 8` |
| `%=` | `i %= 8` | `i = i % 8` |
| `**=` | `i **= 8` | `i = i ** 8` |

```python
graded = 0
graded += 5
print(graded)
```

Because it reads the variable first, the variable must already hold a value.

```python
try:
    ungraded += 1
except NameError as error:
    print("NameError:", error)
```

The operator applies to the whole expression on the right, not just its first term.

```python
i, j = 2, 3
i *= j + 5        # i = i * (3 + 5), not i * 3 + 5
print(i)
```

Simultaneous assignment binds several names in one statement:
`var1, var2 = exp1, exp2`.

```python
midterm, final = 82, 91
print(midterm, final)
```

Every expression on the right is evaluated before any name on the left is bound, which is
why swapping two variables needs no temporary.

```python
midterm, final = final, midterm
print(midterm, final)
```

The counts on the two sides must match.

```python
try:
    a, b, c = 1, 2
except ValueError as error:
    print("ValueError:", error)
```

# 10. Named constants

A named constant is a variable you promise not to reassign. Python has no `const` keyword
and will not stop you changing it, so the all-uppercase name is the entire mechanism: it
tells every reader the value is fixed.

This is correct, and unreadable in a month. What is 0.0825? What is 4.75?

```python
print(189.00 + 189.00 * 0.0825 + 4.75)
```

The same total, with each number named where it is defined and used once.

```python
TEXAS_TAX_RATE = 0.0825
SHIPPING_FLAT_FEE = 4.75

textbook_price = 189.00
tax = textbook_price * TEXAS_TAX_RATE
total = textbook_price + tax + SHIPPING_FLAT_FEE

print(f"tax:   ${tax:.2f}")
print(f"total: ${total:.2f}")
```

The payoff is the second reason to do it: when the rate changes, there is exactly one line
to edit, not a search through the program for every `0.0825`.

# 11. Very large and very small numbers

Scientific notation writes a float as a mantissa times a power of ten, with `e` or `E` for
the power.

| Literal | Means | Value |
|---|---|---|
| `1.23456e2` | 1.23456 x 10^2 | 123.456 |
| `1.23456e+2` | the same; the `+` is optional | 123.456 |
| `1.23456e-2` | 1.23456 x 10^-2 | 0.0123456 |

```python
print(1.23456e2)
print(1.23456e-2)
```

A literal written that way is a `float` even when the value is whole.

```python
print(2e3, type(2e3))
```

Python switches to the same notation on output once the decimal form stops being
readable.

```python
print(0.0001)
print(0.00001)
```

Overflow is a result too large for the type holding it. An `int` has no fixed size, so
integers never overflow.

```python
print(len(str(245 ** 1000)), "digits, exact")
```

A `float` is fixed at 64 bits, with a largest value near 1.8 x 10^308. Past that,
`OverflowError`.

```python
import sys

print(sys.float_info.max)

try:
    245.0 ** 1000
except OverflowError as error:
    print("OverflowError:", error)
```

Arithmetic on values that are already floats gives `inf` instead of raising.

```python
print(1e308 * 10)
```

Underflow is the other end: too close to zero to represent. Python approximates it to `0.0`
and says nothing, so the danger is dividing by the result.

```python
print(1e-320 / 1e10)
```

# 12. Reading the clock

An `import` statement makes a module's contents available under its name, and belongs at the
top of a program. `time.time()` returns the seconds elapsed since midnight, 1 January 1970
Coordinated Universal Time (UTC), a reference point called the epoch.

```python
import time

print(time.time())
```

Turning that count into a clock time is nothing but part 8, applied four times.

```python
total_seconds = int(time.time())

current_second = total_seconds % 60
total_minutes = total_seconds // 60
current_minute = total_minutes % 60
total_hours = total_minutes // 60
current_hour = total_hours % 24

print(current_hour, current_minute, current_second)
```

`:02d` pads a single digit with a leading zero, so 16:4:9 reads as 16:04:09.

```python
print(f"{current_hour:02d}:{current_minute:02d}:{current_second:02d} UTC")
```

Because the value is a float measured in seconds, the same function is a stopwatch: read it
before and after, then subtract.

```python
import math

start = time.time()
value = math.factorial(200_000)
elapsed = time.time() - start

print(f"{elapsed:.4f} seconds")
```

# 13. From problem to program

The software development process is the ordered set of phases that turns a problem statement
into working, maintained software. The phases are the same for a 20-line class exercise and
a system with a thousand users. Only the amount of work in each one changes.

    requirement specification
              |
              v
        system analysis          input -> process -> output
              |
              v
         system design
              |
              v
        implementation
              |
              v
           testing
              |
              v
         deployment  ->  maintenance  ->  back to requirements

| Phase | What it produces |
|---|---|
| Requirement specification | A written statement of what the software must do, agreed with the people who will use it. |
| System analysis | The data flow: what goes in, what comes out, what the system has to know. |
| System design | The components, what each is responsible for, and how they relate. |
| Implementation | Code for each component, written, debugged, and working together. |
| Testing | Evidence that the code meets the requirement specification, ideally gathered by someone who did not write it. |
| Deployment | The software made available to its users. |
| Maintenance | Fixes and changes as the requirements and the environment move. |

The core of analysis and design is IPO: input, process, output. Name the inputs, name the
outputs, then work out the process between them. On a program this small it takes two
minutes on paper, and it is what stops you writing the program three times.

Tracing is the other half: stepping through the statements by hand and writing down what
each variable holds after each one. It is how you find a wrong answer that raises no error.

# 14. Program: monthly loan payment

**Requirement.** Given a loan amount, an annual interest rate, and a term in years, report
the monthly payment and the total amount repaid.

| | |
|---|---|
| Input | annual interest rate as a percent, term in years, loan amount |
| Process | convert the annual percent to a monthly rate, apply the payment formula, multiply by the number of payments |
| Output | monthly payment and total repaid, each to two decimals |

The monthly payment on a fixed-rate loan is

    payment = amount * monthly_rate / (1 - 1 / (1 + monthly_rate) ** months)

**Input.** Literals here so the cells re-run; the commented lines are the live version.

```python
# annual_interest_rate = float(input("Annual interest rate in percent: "))
# years = int(input("Term in years: "))
# loan_amount = float(input("Loan amount: "))
annual_interest_rate = 5.5
years = 10
loan_amount = 27000.00
```

**Process.** Both conversions are where this program goes wrong in practice. An annual rate
of 5.5 percent is `5.5 / 100 / 12` per month, not `5.5 / 12`. A 10-year term is `10 * 12`
payments, not 10. Named constants put both conversions in plain sight.

```python
MONTHS_PER_YEAR = 12
PERCENT = 100

monthly_rate = annual_interest_rate / PERCENT / MONTHS_PER_YEAR
months = years * MONTHS_PER_YEAR

monthly_payment = loan_amount * monthly_rate / (1 - 1 / (1 + monthly_rate) ** months)
total_payment = monthly_payment * months
```

**Output.** Formatted, never raw.

```python
print(f"Loan amount:     ${loan_amount:>10,.2f}")
print(f"Monthly payment: ${monthly_payment:>10,.2f}")
print(f"Total repaid:    ${total_payment:>10,.2f}")
print(f"Interest paid:   ${total_payment - loan_amount:>10,.2f}")
```

Set `years` to 20 and run the three cells again. The monthly payment falls and the interest
paid roughly doubles, which is the argument about loan terms in two numbers.

# 15. Program: distance between two points

**Requirement.** Given two points, report the straight-line distance between them.

| | |
|---|---|
| Input | `x1`, `y1`, `x2`, `y2` |
| Process | the horizontal and vertical differences, then the distance formula |
| Output | the distance, to two decimals |

The distance between (x1, y1) and (x2, y2) is

    ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

**Input and process.** The exponent `0.5` is the square root, so this needs nothing beyond
part 2.

```python
x1, y1 = 1.5, -3.4
x2, y2 = 4.0, 5.9

dx = x2 - x1
dy = y2 - y1

print((dx ** 2 + dy ** 2) ** 0.5)
```

The `math` module spells the same computation two clearer ways. `math.hypot` takes the two
differences directly and stays accurate on coordinates large enough to overflow the
squares, so it is the one to reach for.

```python
import math

print(math.sqrt(dx ** 2 + dy ** 2))
print(math.hypot(dx, dy))
```

**Output.**

```python
print(f"Distance from ({x1}, {y1}) to ({x2}, {y2}) is {math.hypot(dx, dy):.2f}")
```
