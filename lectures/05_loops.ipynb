# 1. Chapter 4 math and string review

Built-in functions need no import. Names that begin with <code>math.</code>

| Examples | Import |

```python
print(max(10, 2))
print(min(10, 2))
print(max(min(max(10, 2), 2), 8))
print(pow(min(2, 3), 3))
```

<code>round</code> uses ties-to-even for exact halfway values, while
<code>int</code> discards the fractional part toward zero.

```python
print(round(4.5), round(5.4), round(5.5))
print(int(4.5), int(5.4), int(5.5))
```

The <code>math</code> module provides rounding directions, square roots, and

```python
import math

print(math.ceil(2.1))
print(math.floor(2.1))
print(math.sqrt(9))
print(f"{math.pi:.4f}")
```

<code>ord</code> converts one character to its Unicode code point;
<code>chr</code> performs the reverse conversion.

```python
print(ord("a"))
print(chr(98))
print(ord("A"))
```

Strings support concatenation with <code>+</code>, repetition with
<code>*</code>, and case-sensitive membership tests with <code>in</code>
and <code>not in</code>.

```python
print("price = $" + str(9.99) + "\n" + "~" * 15)
print("python" in "programming using python")
print("python" not in "programming using PYTHON")
```

Indexes select one character. A slice includes its start index and excludes

```python
text = "ABCabc"
print(text[4], text[0], text[-1])
print(text[1:3])
print(text[:2] + " " + text[2:-1])
print(len("COSC 1436"))
print(min("ABCabc"))
```

String comparisons are lexicographic and case-sensitive. Quote style does
not change a string's value.

```python
print("Hey there" > "hi")
print('Python' == "Python")

first = "argentina"
second = "Zimbabwe"
if second < first:
    first, second = second, first
print("Smallest string first:", first, second)
```

Content-testing methods describe the characters in a non-empty string.

```python
print("1436".isdigit())
print("COSC1436".isalpha())
print("COSCXXXX".isalpha())
print("COSC1436".isalnum())
print("COSC XXXX".isupper())
print("COSC XXXX".islower())
print("COSC XXXX".isspace())
print("      ".isspace())
```

Search methods report positions, counts, prefixes, and suffixes. The correct

Conversion methods return new strings. Strip methods remove characters from

```python
print("welcome".capitalize())
print("cosc".upper())
print("WELCOME".lower())
print(repr("    welcome    ".strip()))
print(repr("    welcome    ".rstrip()))
print(repr("    welcome".lstrip()))
```

A format specifier controls width, alignment, and precision. F-strings put

```python
print("\"Knowledge\"", end=" ")
print("is the key to 'success'")
print("name\tvalue")
print("\n\"\\\\hello there!\"")
```

# 2. A while loop repeats while its condition is true

Repeating a statement by hand makes maintenance difficult. A loop states the
repeated action once.

```python
print("Programming is fun!")
print("Programming is fun!")
print("Programming is fun!")
```

A counter-controlled <code>while</code> loop has four parts: initialize the
counter, test a condition, run the body, and update the counter.

<strong>initialize → test → body → update ↺</strong>

```python
count = 0
while count < 3:
    print("Programming is fun!")
    count += 1
```

Trace the state before trusting the loop.

| Start count | <code>count &lt; 3</code> | Printed | Next count |
|---:|:---:|---|---:|
| 0 | True | Programming is fun! | 1 |
| 1 | True | Programming is fun! | 2 |
| 2 | True | Programming is fun! | 3 |
| 3 | False | nothing | stop |

The loop-control variable keeps its final value after the loop.

```python
count = 0
while count < 3:
    count += 1
print(count)
```

A false condition at the first test means the body runs zero times.

```python
count = 3
ran = False
while count < 3:
    ran = True
print(ran)
```

The update must make progress toward a false condition. This loop counts
downward, so its test and update point in the same direction.

```python
number = 6
while number > 0:
    print(number, end=" ")
    number -= 2
print("done")
```

Boundary choices create common off-by-one errors. Use <code>&lt;</code> when
the upper boundary is excluded and <code>&lt;=</code> when it is included.

```python
number = 1
while number <= 3:
    print(number, end=" ")
    number += 1
print()
```

Never run a loop whose state cannot change. The following safe demonstration
includes the missing update that an infinite version would omit.

```python
attempts_left = 3
while attempts_left > 0:
    print(f"{attempts_left} attempt(s) left")
    attempts_left -= 1
```

# 3. Design the state before writing the loop

Answer four questions before coding:

1. What state exists before the first test?
2. What condition allows another iteration?
3. What useful work happens in the body?
4. What update guarantees progress?

An accumulator stores a result that grows over time. This loop adds the
integers from 1 through 5.

```python
total = 0
number = 1
while number <= 5:
    total += number
    number += 1
print(total)
```

A number divisible by 2, 3, and 5 is divisible by 30. Stepping by 30 avoids
testing values that cannot qualify.

```python
total = 0
number = 30
while number < 1000:
    total += number
    number += 30
print(total)
```

Strings are immutable, but a loop can build a new string. This countdown
includes both boundaries.

```python
upper_bound = 9
lower_bound = 5
countdown = ""

number = upper_bound
while number >= lower_bound:
    countdown += str(number)
    number -= 1

print(countdown)
```

A loop can improve an estimate until it reaches a goal. Here the estimate
approaches the secret in fixed steps, so termination is predictable.

```python
secret = 42
guess = 0
attempts = 0

while guess != secret:
    guess += 7
    attempts += 1
    print(f"Attempt {attempts}: {guess}")

print("Found it.")
```

# 4. Sentinels and user-controlled repetition

A sentinel is a special value that ends an otherwise open-ended sequence.
The sentinel is tested but is not included in the calculation.

```python
data = "5720"
position = 0
total = 0
value = int(data[position])

while value != 0:
    total += value
    position += 1
    value = int(data[position])

print(total)
```

A confirmation loop continues only while the response means yes. This
deterministic example changes the response after three passes so Run All
never pauses for keyboard input.

```python
response = "y"
pass_number = 0

while response.lower() == "y":
    pass_number += 1
    print(f"Pass {pass_number}")
    response = "n" if pass_number == 3 else "y"
```

Interactive programs can read redirected input without changing their
Python code:

    python program.py < input.txt
    python program.py > output.txt
    python program.py < input.txt > output.txt

Redirection belongs to the shell. The program still uses ordinary
<code>input</code> and <code>print</code> calls.

# 5. for loops and range

Use <code>for</code> when iterating through known values. It handles the
loop-control update automatically.

```python
for number in range(3):
    print(number)
```

<code>range(stop)</code> starts at 0 and excludes <code>stop</code>.

```python
for number in range(4):
    print(number, end=" ")
print()
```

<code>range(start, stop)</code> includes <code>start</code> and excludes
<code>stop</code>.

```python
for number in range(4, 8):
    print(number, end=" ")
print()
```

A positive step skips upward through the sequence.

```python
for number in range(3, 9, 2):
    print(number, end=" ")
print()
```

A negative step counts downward. The stop value is still excluded.

```python
for number in range(5, 1, -1):
    print(number, end=" ")
print()
```

A string is iterable, so a <code>for</code> loop can visit one character at
a time.

```python
text = "Community College"
vowel_count = 0

for character in text.lower():
    if character in "aeiou":
        vowel_count += 1

print(vowel_count)
```

Prepending each new character builds a reversed copy of a string.

```python
original = "Community College"
reversed_text = ""

for character in original:
    reversed_text = character + reversed_text

print(reversed_text)
```

A counter and accumulator work together to calculate an average. This
version keeps positive integers below 1000 that are divisible by none of
2, 5, or 7.

```python
total = 0
count = 0

for number in range(1, 1000):
    if number % 2 != 0 and number % 5 != 0 and number % 7 != 0:
        total += number
        count += 1

average = total / count
print(f"{average:.2f}")
```

Choose by intent: <code>while</code> answers “repeat until this state
changes,” while <code>for</code> answers “visit each value in this
sequence.”

# 6. Nested loops coordinate repeated dimensions

The inner loop completes all of its iterations for each single outer-loop
iteration.

```python
for row in range(1, 3):
    for column in range(1, 4):
        print(f"({row}, {column})", end=" ")
    print()
```

A growing upper boundary creates a triangular pattern.

```python
for row in range(1, 7):
    for value in range(1, row + 1):
        print(value, end=" ")
    print()
```

A multiplication table is a natural row-by-column problem.

```python
print("     1   2   3   4   5")
for row in range(1, 6):
    print(f"{row:>2}:", end="")
    for column in range(1, 6):
        print(f"{row * column:>4}", end="")
    print()
```

# 7. Keep floating-point equality out of loop control

Many decimal fractions have no exact binary floating-point representation.
A value that looks like 0.3 may not equal the stored result of repeated
addition.

```python
total = 0.1 + 0.1 + 0.1
print(repr(total))
print(total == 0.3)
```

Repeated subtraction may pass close to zero without ever becoming exactly
zero. The iteration guard keeps this demonstration finite.

```python
item = 1.0
iterations = 0

while item != 0 and iterations < 15:
    item -= 0.1
    iterations += 1

print(repr(item))
print(iterations)
```

Count exact integer units in the loop, then convert to a float only for
display. This sums 0.01 through 1.00 without a floating-point loop counter.

```python
total_cents = 0
for cents in range(1, 101):
    total_cents += cents

print(f"{total_cents / 100:.2f}")
```

When approximate comparison is the real intent, use
<code>math.isclose</code> instead of equality.

```python
print(math.isclose(0.1 + 0.1 + 0.1, 0.3))
```

# 8. break exits and continue skips

<code>break</code> exits the nearest loop immediately. It is useful after a
search has found what it needs.

```python
for number in range(22, 100):
    if number % 7 == 0:
        first_multiple = number
        break

print(first_multiple)
```

<code>continue</code> skips the rest of the current iteration and begins the
next one.

```python
for number in range(1, 10):
    if number in (3, 4):
        continue
    print(number, end=" ")
print()
```

A guessing search can stop as soon as the generated guess matches the
target.

```python
secret = 42
attempts = 0

for guess in range(0, 101, 7):
    attempts += 1
    if guess == secret:
        print(f"Found {secret} in {attempts} attempts.")
        break
```

Use these statements deliberately: a clear loop condition is often easier
to trace than several early exits or skips.

# 9. Loop case studies

The Euclidean algorithm repeatedly replaces two integers with the smaller
integer and the remainder. It stops when the remainder becomes zero.

```python
first = 84
second = 30

while second != 0:
    first, second = second, first % second

print(first)
```

A condition-controlled loop can answer “how long until?” questions. This
example finds when $10,000 tuition first reaches twice its starting value
with 7% annual growth.

```python
tuition = 10000.0
target = tuition * 2
years = 0

while tuition < target:
    tuition *= 1.07
    years += 1

print(f"{years} years: $" + f"{tuition:,.2f}")
```

Repeated division by 16 produces hexadecimal digits from right to left.
Prepending each remainder produces the final order.

```python
decimal = 48879
digits = "0123456789ABCDEF"
value = decimal
hexadecimal = "0" if value == 0 else ""

while value > 0:
    remainder = value % 16
    hexadecimal = digits[remainder] + hexadecimal
    value //= 16

print(f"{decimal} = {hexadecimal}")
```

A palindrome check moves inward from both ends and stops on a mismatch or
when the indexes meet.

```python
phrase = "Never odd or even"
compact = ""

for character in phrase:
    if character.isalnum():
        compact += character.lower()

left = 0
right = len(compact) - 1
is_palindrome = True

while left < right and is_palindrome:
    if compact[left] != compact[right]:
        is_palindrome = False
    left += 1
    right -= 1

print(is_palindrome)
```

Prime generation combines an outer candidate loop with an inner divisor
test. Testing only while <code>divisor * divisor &lt;= number</code> is
sufficient.

```python
number = 2
prime_count = 0

while prime_count < 20:
    divisor = 2
    is_prime = True

    while divisor * divisor <= number and is_prime:
        if number % divisor == 0:
            is_prime = False
        divisor += 1

    if is_prime:
        print(f"{number:>3}", end=" ")
        prime_count += 1
        if prime_count % 10 == 0:
            print()

    number += 1
```

A seeded one-dimensional random walk is reproducible: each step moves left
or right by one position.

```python
import random

random.seed(1436)
position = 0

for step in range(1, 13):
    direction = -1 if random.randint(0, 1) == 0 else 1
    position += direction
    print(f"step {step:>2}: position {position:>2}")
```

# 10. Program: text analysis report

<strong>Requirement.</strong> Normalize a phrase, count its character
categories, build a compact alphanumeric form, reverse the original text,
test for a palindrome, and print a formatted report.

Store the phrase that the program will analyze.

```python
raw_text = "A1 b2 B1a"
```

Normalize case and remove whitespace only from the two ends.

```python
text = raw_text.strip().lower()
```

Initialize every counter and every string accumulator.

```python
letters = 0
digits = 0
spaces = 0
vowels = 0
compact = ""
reversed_text = ""
```

Visit each character once. The loop classifies the character, builds the
compact form, and constructs the reverse.

```python
for character in text:
    reversed_text = character + reversed_text

    if character.isalpha():
        letters += 1
        if character in "aeiou":
            vowels += 1
    elif character.isdigit():
        digits += 1
    elif character.isspace():
        spaces += 1

    if character.isalnum():
        compact += character
```

Compare matching positions from the outside toward the center.

```python
left = 0
right = len(compact) - 1
is_palindrome = True

while left < right and is_palindrome:
    if compact[left] != compact[right]:
        is_palindrome = False
    left += 1
    right -= 1
```

Use aligned f-strings to present the completed report.

```python
print(f"{'Original:':<18}{raw_text}")
print(f"{'Normalized:':<18}{text}")
print(f"{'Reversed:':<18}{reversed_text}")
print(f"{'Compact:':<18}{compact}")
print(f"{'Letters:':<18}{letters}")
print(f"{'Digits:':<18}{digits}")
print(f"{'Spaces:':<18}{spaces}")
print(f"{'Vowels:':<18}{vowels}")
print(f"{'Palindrome:':<18}{is_palindrome}")
```

The final program combines Chapter 4 string tools with Chapter 5 repetition:
<code>for</code> handles a known sequence of characters, while
<code>while</code> handles a condition that changes as two indexes move.
