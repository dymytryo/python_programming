# List is a Sequence

| **Operation**       | **Description**                                                                 |
|---------------------|---------------------------------------------------------------------------------|
| `x (not) in s`            | True if element `x` is (not) in sequence `s`.                                        |
| `s1 + s2`           | Concatenates two sequences `s1` and `s2`.                                       |
| `s * n`, `n * s`    | `n` copies of sequence `s` concatenated.                                        |
| `s[i]`              | `i`th element in sequence `s`.                                                  |
| `s[i : j]`          | Slice of sequence `s` from index `i` to `j–1`.                                  |
| `len(s)`            | Length of sequence `s`, i.e., the number of elements in `s`.                    |
| `min(s)`            | Smallest element in sequence `s`.                                               |
| `max(s)`            | Largest element in sequence `s`.                                                |
| `sum(s)`            | Sum of all numbers in sequence `s`.                                             |
| `for` loop          | Traverses elements from left to right in a `for` loop.                          |
| `<`, `<=`, `>`, `>=`, `==`, `!=` | Compares two sequences.                                               |

<b>Assignment</b>: Given the list of numbers numbers, write some code that:

- assigns to <code>minimum</code> the minimum value;
- assigns to <code>maximum</code> the maximum value;
- assigns to <code>maxOffset</code> the difference between the maximum and average values;
- assigns to <code>minOffset</code> the difference between the minimum and average values.
Note: the Offset variables should be absolute values.
</div>

```python
numbers = [10, 20, 30, 40, 50]

minimum = min(numbers)
maximum = max(numbers)
average = sum(numbers) / len(numbers)

maxOffset = abs(maximum - average)
minOffset = abs(minimum - average)

print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Max Offset: {maxOffset}")
print(f"Min Offset: {minOffset}")
```

| **Method**             | **Description**                                                                                  |
|------------------------|--------------------------------------------------------------------------------------------------|
| `append(x)`            | Adds an element `x` to the end of the list.                                                      |
| `count(x)`             | Returns the number of times element `x` appears in the list.                                     |
| `extend(anotherList)`  | Appends all the elements in `anotherList` to the list.                                           |
| `index(x)`             | Returns the index of the first occurrence of element `x` in the list.                            |
| `insert(index, x)`     | Inserts an element `x` at a given `index`. Note that the first element in the list has index `0`.|
| `pop(index)`           | Removes the element at the given `index` and returns it. If not specified, removes and returns the last element. |
| `remove(x)`            | Removes the first occurrence of element `x` from the list.                                       |
| `reverse()`            | Reverses the elements in the list.                                                               |
| `sort()`               | Sorts the elements in the list in ascending order.                                               |

<div style="background-color:#FAF9F6; color:#283618; padding:10px; border-radius:5px;">
<b>Assignment</b>: Assume two lists list1 and list2, already declared and filled. Write a snippet of code that interleaves the two lists, and assigns the result to a list interleaved.

For example:

list1 = [1, 2, 3]
list2=['a', 'b', 'c']
interleaved=[1, 'a', 2, 'b', 3, 'c']
Note: Your code should take into account cases where the two lists are of different lengths. In that case, the interleaving will be executed as usual, and any remaining elements of the longer list will be appended to the end of interleaved.
</div>

```python
list1 = [1, 2, 3] 
list2=['a', 'b', 'c', 'd', 'e']
interleaved = []
i = 0

# Loop through both lists until one of them ends
while i < len(list1) and i < len(list2):
    interleaved.append(list1[i])
    interleaved.append(list2[i])
    i += 1

# Add any remaining elements from list1
while i < len(list1):
    interleaved.append(list1[i])
    i += 1

# Add any remaining elements from list2
while i < len(list2):
    interleaved.append(list2[i])
    i += 1

interleaved
```
